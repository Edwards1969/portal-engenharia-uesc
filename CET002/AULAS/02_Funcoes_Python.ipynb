  1. <strong>Introdução às funções em Python</strong>
</div>

Uma **função** reúne instruções que realizam uma tarefa específica. Em Engenharia, podemos criar uma função para calcular uma vazão, converter a leitura de um sensor ou determinar um erro de controle. Depois de definida, a função pode ser utilizada diversas vezes, com diferentes dados.

A palavra-chave `def` inicia a definição de uma função. **Definir** não significa **executar**: a execução acontece quando chamamos a função pelo nome.

<div id="secao-2" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  2. Funções nativas do Python (<em>built-in functions</em>)
</div>

Antes de construir funções com `def`, vale reconhecer que já utilizamos funções prontas do Python. `print()` apresenta informações na tela e `type()` identifica o tipo de um objeto. Outras funções nativas realizam operações úteis com dados de instrumentos e cálculos de Engenharia. **Não precisamos usar `def` para chamar essas funções.**

| Função | O que faz | Exemplo de aplicação |
|---|---|---|
| `abs(x)` | Retorna o valor absoluto | Magnitude do erro de controle |
| `round(x, n)` | Arredonda para `n` casas decimais | Apresentação de medidas |
| `min(valores)` | Obtém o menor valor | Menor temperatura registrada |
| `max(valores)` | Obtém o maior valor | Maior pressão registrada |
| `sum(valores)` | Soma os valores | Totalização de volumes por intervalos |

Vamos começar com um cálculo de Controle de Processos. Para um sistema de temperatura, o erro é $e=T_{SP}-T_{PV}$. Se a referência é $80\ ^\circ\mathrm{C}$ e a leitura é $84\ ^\circ\mathrm{C}$, então $e=80-84=-4\ ^\circ\mathrm{C}$. O sinal indica o sentido do desvio; a magnitude é $|e|=4\ ^\circ\mathrm{C}$.

```python
T_SP = 80     # °C
T_PV = 84     # °C

erro = T_SP - T_PV
magnitude = abs(erro)

print("Erro com sinal:", erro, "°C")
print("Magnitude do erro:", magnitude, "°C")
```

**Interpretação:** `abs()` não substitui o erro com sinal em uma lei de controle. Ela é útil quando queremos saber *quanto* a medida se afastou da referência, independentemente da direção.

Agora considere cinco temperaturas registradas por um instrumento: $[72{,}4; 73{,}1; 71{,}8; 74{,}2; 72{,}9]\ ^\circ\mathrm{C}$. Podemos identificar os extremos e calcular a média usando `min()`, `max()` e `sum()`. A média aritmética é $\bar T=\frac{\sum_{i=1}^{5}T_i}{5}=72{,}88\ ^\circ\mathrm{C}$. A função `round()` será usada apenas para apresentar a média com uma casa decimal.

```python
temperaturas = [72.4, 73.1, 71.8, 74.2, 72.9]

menor = min(temperaturas)
maior = max(temperaturas)
media = sum(temperaturas) / len(temperaturas)

print("Menor temperatura:", menor, "°C")
print("Maior temperatura:", maior, "°C")
print("Temperatura média:", round(media, 1), "°C")
```

**Atenção:** `round()` é útil para a apresentação, mas convém manter os valores sem arredondamento prematuro durante cálculos sucessivos. Além disso, `sum()` soma números: para totalizar volume a partir de vazões, é necessário considerar o intervalo de tempo e as unidades.

<div id="secao-3" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  3. <strong>Criando a primeira função com def</strong>
</div>

Vamos começar com uma função que apenas apresenta uma mensagem. Observe os dois pontos (`:`) e a indentação: as instruções recuadas pertencem à função.

```python
def apresentar():
    print("Iniciando os cálculos de Engenharia")

apresentar()
```

A instrução `def apresentar():` cria a função. A chamada `apresentar()` executa o que está dentro dela. Não precisamos utilizar `return` quando a finalidade é apenas mostrar uma mensagem.

<div id="secao-4" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  4. <strong>Funções com parâmetros e argumentos</strong>
</div>

Em Instrumentação, uma temperatura pode ser medida em graus Celsius e convertida para kelvin. O modelo matemático é

$$T_K=T_{°C}+273{,}15.$$

O valor recebido pela função é chamado de **parâmetro**; o valor fornecido na chamada é o **argumento**.

```python
def converter_kelvin(temperatura_c):
    temperatura_k = temperatura_c + 273.15
    return temperatura_k

print(converter_kelvin(25))
```

Cálculo manual: para $25\ ^\circ\mathrm{C}$, $T_K=25+273{,}15=298{,}15\ \mathrm{K}$. Na definição, `temperatura_c` é o parâmetro; na chamada, `25` é o argumento.

<div id="secao-5" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  5. <strong>A diferença entre print() e return</strong>
</div>

`print()` **mostra** informações na tela; `return` **devolve** um valor ao programa. O valor devolvido pode ser armazenado e utilizado em outros cálculos. Além disso, `return` encerra a execução da função.

```python
def erro_temperatura(referencia, medida):
    return referencia - medida

erro = erro_temperatura(150, 143)
print("Erro de temperatura:", erro, "°C")
print("Módulo do erro:", abs(erro), "°C")
```

No controle de temperatura, o erro é $e=T_{SP}-T_{PV}$. Para $T_{SP}=150\ ^\circ\mathrm{C}$ e $T_{PV}=143\ ^\circ\mathrm{C}$, temos $e=7\ ^\circ\mathrm{C}$. A variável `erro` recebe o valor retornado pela função.

<div id="secao-6" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  6. <strong>Funções com múltiplos parâmetros</strong>
</div>

Uma função pode receber vários parâmetros. Na Mecânica dos Fluidos, a vazão volumétrica é $Q=A\,v$, em que $A$ é a área da seção transversal (m²) e $v$ é a velocidade média (m/s).

```python
def vazao_volumetrica(area, velocidade):
    return area * velocidade

Q = vazao_volumetrica(0.02, 1.5)
print("Vazão volumétrica:", Q, "m³/s")
```

Cálculo manual: $Q=0{,}02\times1{,}5=0{,}03\ \mathrm{m^3/s}$. Experimente alterar apenas a velocidade e executar novamente.

<div id="secao-7" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  7. <strong>Funções que retornam múltiplos resultados</strong>
</div>

Às vezes precisamos de mais de uma informação. Para uma corrente líquida, podemos calcular a vazão volumétrica $Q=A\,v$ e a vazão mássica $\dot{m}=\rho Q$. A função pode devolver ambos os resultados.

```python
def calcular_vazoes(area, velocidade, rho):
    Q = area * velocidade
    m_dot = rho * Q
    return Q, m_dot

Q, m_dot = calcular_vazoes(0.02, 1.5, 1000)
print("Vazão volumétrica:", Q, "m³/s")
print("Vazão mássica:", m_dot, "kg/s")
```

Cálculo manual: $Q=0{,}03\ \mathrm{m^3/s}$ e $\dot{m}=1000\times0{,}03=30\ \mathrm{kg/s}$. Python reúne os valores retornados em uma tupla, que pode ser desempacotada nas variáveis `Q` e `m_dot`.

<div id="secao-8" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  8. <strong>Utilizando funções com listas e laços for</strong>
</div>

Agora vamos reunir duas ferramentas: o laço `for`, que percorre os dados, e a função, que realiza um cálculo. Em um experimento, registramos temperaturas de um fluido em diferentes instantes e queremos convertê-las para kelvin.

```python
def converter_kelvin(temperatura_c):
    return temperatura_c + 273.15

temperaturas_c = [20, 25, 30, 35]
temperaturas_k = []

for T in temperaturas_c:
    temperaturas_k.append(converter_kelvin(T))

print(temperaturas_k)
```

Observe que `append()` guarda cada resultado na nova lista. O `return` pertence à função `converter_kelvin`, enquanto o `for` controla a repetição. **Exercício:** inclua a temperatura de $40\ ^\circ\mathrm{C}$ na lista e execute novamente.

<div id="secao-9" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  9. <strong>Funções aplicadas à Instrumentação e Controle de Qualidade</strong>
</div>

**Instrumentação — conversão de sinal de um transmissor de temperatura.** Considere um transmissor ideal, linear, de 4–20 mA, calibrado para 0–200 °C. Para uma corrente $I$ (mA), a temperatura é

$$T=\frac{I-4}{20-4}\,(200-0).$$

A equação é válida para o intervalo nominal de 4 a 20 mA.

Cálculo manual: para $I=12\ \mathrm{mA}$, $T=(12-4)/16\times200=100\ ^\circ\mathrm{C}$.

```python
def corrente_para_temperatura(corrente_ma):
    return (corrente_ma - 4) / 16 * 200

for corrente in [4, 8, 12, 16, 20]:
    print(corrente, "mA ->", corrente_para_temperatura(corrente), "°C")
```

**Interpretação:** a função transforma o sinal elétrico em uma grandeza física. Em aplicações reais, leituras fora da faixa nominal exigem avaliação de falha, saturação ou condição anormal; não devem ser interpretadas automaticamente como temperaturas válidas.

<div id="secao-10" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  10. <strong>Funções aplicadas ao Controle de Processos</strong>
</div>

**Controle de Processos — ação proporcional.** Um controlador proporcional ideal pode ser descrito, de forma simplificada, por $u=K_p e$, em que $e=r-y$ é o erro entre referência e medida. Neste exemplo, $u$ representa apenas a parcela proporcional da ação de controle, sem considerar polarização, saturação ou dinâmica do atuador.

Cálculo manual: com $r=80\ ^\circ\mathrm{C}$, $y=76\ ^\circ\mathrm{C}$ e $K_p=2$, temos $e=4\ ^\circ\mathrm{C}$ e $u=8$ unidades de ação por convenção do exemplo.

```python
def controle_proporcional(referencia, medida, Kp):
    erro = referencia - medida
    acao = Kp * erro
    return erro, acao

erro, acao = controle_proporcional(80, 76, 2)
print("Erro:", erro, "°C")
print("Ação proporcional:", acao)
```

  11. <strong>Funções aplicadas à Mecânica dos Fluidos e Transferência de Calor</strong>
</div>

**Mecânica dos Fluidos — velocidade média em uma tubulação circular.** A área da seção é $A=\pi D^2/4$ e a velocidade média é $v=Q/A$, para uma vazão volumétrica $Q$.

Cálculo manual: para $D=0{,}10\ \mathrm{m}$ e $Q=0{,}015\ \mathrm{m^3/s}$, temos $A\approx0{,}007854\ \mathrm{m^2}$ e $v\approx1{,}91\ \mathrm{m/s}$.

```python
import math

def velocidade_media(Q, diametro):
    area = math.pi * diametro**2 / 4
    return Q / area

print("Velocidade média:", round(velocidade_media(0.015, 0.10), 2), "m/s")
```

**Transferência de Calor — condução unidimensional em parede plana.** Em regime permanente, para condutividade constante e sem geração interna, o módulo da taxa de calor é $\dot{Q}=kA(T_1-T_2)/L$, quando $T_1>T_2$.

Cálculo manual: com $k=0{,}8\ \mathrm{W/(m\,K)}$, $A=2\ \mathrm{m^2}$, $T_1=80\ ^\circ\mathrm{C}$, $T_2=30\ ^\circ\mathrm{C}$ e $L=0{,}10\ \mathrm{m}$, resulta $\dot{Q}=800\ \mathrm{W}$.

```python
def taxa_conducao(k, area, T1, T2, espessura):
    return k * area * (T1 - T2) / espessura

print("Taxa de calor:", taxa_conducao(0.8, 2, 80, 30, 0.10), "W")
```

A função retorna uma taxa com sinal associado à ordem escolhida para as temperaturas. Neste exemplo, o resultado positivo corresponde ao fluxo do lado mais quente para o mais frio.
