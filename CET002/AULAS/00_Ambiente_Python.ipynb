Python é uma linguagem de programação de alto nível, com sintaxe relativamente simples e recursos para diferentes estilos de programação. Na Engenharia, sua principal utilidade não é substituir o raciocínio físico e matemático, mas **transformar modelos e cálculos em procedimentos reproduzíveis**.

Considere, por exemplo, a determinação da vazão mássica de um fluido. Primeiro identificamos a relação física $\dot m=\rho Q$. Depois informamos os valores de massa específica e vazão volumétrica. Finalmente, utilizamos Python para calcular e conferir o resultado. Esse caminho — **fenômeno físico → modelo matemático → algoritmo → código → interpretação** — será adotado neste material.

Bibliotecas como NumPy, SciPy, pandas e Matplotlib serão úteis em etapas posteriores para cálculos numéricos, simulações, análise de dados experimentais e gráficos. Entretanto, começaremos com os recursos básicos da própria linguagem, para compreender o que cada instrução realiza.

<div id="secao-2" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">2. Instalação do Anaconda, VS Code e Jupyter Notebook</div>

Um **ambiente virtual**, por sua vez, é um espaço isolado de instalação de pacotes e do interpretador, útil para evitar conflitos entre projetos. Ele não é sinônimo de notebook ou IDE.

Para esta aula, uma opção é instalar o Anaconda e abrir o Jupyter Notebook. Outra é utilizar Python e VS Code com as extensões Python e Jupyter. O aluno não precisa instalar todas as ferramentas simultaneamente.

__4.2. Operações matemáticas__

Considere $a=9$ e $b=4$. Antes de executar, tente antecipar os resultados. A divisão `/` calcula o quociente usual; `//` realiza divisão por piso (arredonda o quociente para baixo); `%` fornece o resto associado a essa divisão; e `**` representa potência. Para números positivos, a divisão por piso coincide com a parte inteira do quociente, mas isso **não vale como regra geral para números negativos**.

````python
a = 9
b = 4

print("Adição:", a + b)
print("Subtração:", a - b)
print("Multiplicação:", a * b)
print("Divisão:", a / b)
print("Divisão por piso:", a // b)
print("Resto:", a % b)
print("Potência:", a ** b)
````

Confira manualmente: $9/4=2{,}25$, $9//4=2$ e $9\%4=1$, pois $9=4\times2+1$. A potência $9^{4}$ vale $6561$. O símbolo `%` **não significa porcentagem** nesse contexto: é o operador de resto da divisão.

__4.3. Aplicações em Engenharia__

**Exemplo 1 — Mecânica dos Fluidos: vazão mássica.** Em um escoamento, a vazão mássica pode ser determinada por

$$\dot m=\rho Q,$$

em que $\rho$ é a massa específica (kg/m³) e $Q$ a vazão volumétrica (m³/s). Para água com $\rho=1000\ \mathrm{kg/m^3}$ e $Q=0{,}015\ \mathrm{m^3/s}$, o cálculo manual fornece

$$\dot m=1000\times0{,}015=15\ \mathrm{kg/s}.$$

````python
rho = 1000      # kg/m³
Q = 0.015        # m³/s
vazao_massica = rho * Q
print("Vazão mássica:", vazao_massica, "kg/s")
````

**Interpretação:** o escoamento transporta 15 kg de água por segundo, nas condições adotadas. Observe que o programa reproduz uma relação física previamente estabelecida.

**Exemplo 2 — Instrumentação e Controle: erro de temperatura.** Em um sistema de controle, definimos o erro como

$$e=T_{SP}-T_{PV}.$$

Se a temperatura desejada é $150\ ^\circ\mathrm C$ e a medida é $142\ ^\circ\mathrm C$, então $e=8\ ^\circ\mathrm C$. O sinal do erro depende da convenção adotada; aqui, utilizamos referência menos medida.

````python
T_SP = 150     # °C: referência
T_PV = 142     # °C: temperatura medida
erro = T_SP - T_PV
print("Erro de temperatura:", erro, "°C")
````

**Interpretação:** a temperatura medida está 8 °C abaixo do valor de referência. Esse cálculo simples será utilizado posteriormente no estudo de controladores, mas aqui o objetivo é compreender variáveis, subtração e saída de resultados.

**Exemplo 3 — Transferência de Calor: condução em parede plana.** Para uma parede homogênea, em regime permanente, com condução unidimensional e condutividade constante, a magnitude da taxa de calor é

$$\dot Q=\frac{kA(T_1-T_2)}{L}.$$

Adotando $k=0{,}8\ \mathrm{W/(m\,K)}$, $A=2\ \mathrm{m^2}$, $T_1=80\ ^\circ\mathrm C$, $T_2=30\ ^\circ\mathrm C$ e $L=0{,}1\ \mathrm m$, obtemos $\dot Q=800\ \mathrm W$. Diferenças de temperatura em °C e K possuem o mesmo valor numérico.

````python
k = 0.8       # W/(m.K)
A = 2.0       # m²
T1 = 80       # °C
T2 = 30       # °C
L = 0.1       # m

Q_calor = k * A * (T1 - T2) / L
print("Taxa de transferência de calor:", Q_calor, "W")
````
