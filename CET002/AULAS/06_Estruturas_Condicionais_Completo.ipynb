  1. Por que um programa precisa tomar decisões?
</div>

Em Engenharia, uma mesma equação nem sempre deve ser utilizada para todas as condições. Um instrumento pode indicar uma grandeza dentro ou fora de sua faixa; uma tubulação pode operar em diferentes regimes de escoamento; um processo pode exigir ações distintas conforme a temperatura. **Uma condição lógica é uma pergunta cuja resposta é `True` (verdadeira) ou `False` (falsa).** O programa usa essa resposta para escolher quais instruções executar.

Neste notebook, estudaremos primeiro a decisão simples, depois alternativas e classificações, e finalmente exemplos que combinam funções e laços `for` dos notebooks anteriores.

```python
temperatura = 82.0
limite = 80.0

print(temperatura > limite)
print(temperatura == limite)
print(temperatura <= limite)
```

O operador `=` atribui um valor; `==` compara dois valores. Essa distinção é essencial.

<div id="secao-2" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  2. A estrutura if: verificando uma condição
</div>

A estrutura `if` executa um bloco **somente quando a condição é verdadeira**. Os dois-pontos (`:`) anunciam o bloco, e a indentação (normalmente quatro espaços) identifica as instruções que pertencem a ele.

**Problema:** um sensor registra a temperatura de um tanque. Queremos emitir uma mensagem se ela ultrapassar um limite didático de 85 °C.

```python
T = 89.0       # °C
T_limite = 85.0 # °C

if T > T_limite:
    print("Atenção: temperatura acima do limite definido.")

print("Verificação concluída.")
```

Experimente substituir `T = 89.0` por `T = 82.0`. A mensagem de atenção desaparece, mas a mensagem final continua sendo exibida: ela está fora do bloco `if`.

<div id="secao-3" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  3. if e else: escolhendo entre dois caminhos
</div>

Quando precisamos escolher entre **duas possibilidades**, usamos `if` e `else`. Se a condição for verdadeira, o primeiro bloco é executado; caso contrário, executa-se o segundo.

**Instrumentação:** suponha um transmissor cuja faixa nominal de corrente seja de 4 a 20 mA. Vamos verificar apenas se uma leitura está dentro dessa faixa; isso não constitui diagnóstico completo de falhas do instrumento.

```python
corrente = 12.4  # mA

if 4.0 <= corrente <= 20.0:
    print("Leitura dentro da faixa nominal de 4–20 mA.")
else:
    print("Leitura fora da faixa nominal: investigar.")
```

**Interpretação de Engenharia.** A comparação encadeada `4.0 <= corrente <= 20.0` verifica simultaneamente os limites inferior e superior. Valores fora da faixa nominal exigem investigação; não devemos atribuir automaticamente uma causa.

<div id="secao-4" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  4. if, elif e else: classificando situações
</div>

Quando há **mais de duas situações**, usamos `if`, um ou mais `elif` (*else if*) e, opcionalmente, `else`. O Python avalia as condições de cima para baixo e executa **apenas o primeiro bloco cuja condição for verdadeira**.

**Controle de temperatura:** vamos classificar a temperatura medida de um processo usando faixas ilustrativas, escolhidas apenas para praticar programação.

```python
T = 78.0  # °C

if T < 70.0:
    estado = "Abaixo da faixa de referência"
elif T <= 85.0:
    estado = "Na faixa de referência"
else:
    estado = "Acima da faixa de referência"

print(estado)
```

Observe que o `elif` não precisa repetir `T >= 70.0`: essa possibilidade já está implícita porque a primeira condição foi falsa. As fronteiras importam: 70 e 85 °C pertencem à faixa intermediária.

<div id="secao-5" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  5. Operadores relacionais e lógicos
</div>

Os operadores relacionais incluem `>`, `<`, `>=`, `<=`, `==` e `!=`. Para combinar condições, utilizamos `and` (ambas verdadeiras), `or` (pelo menos uma verdadeira) e `not` (negação).

**Controle de qualidade:** uma amostra hipotética será considerada dentro de duas especificações simultâneas: pH entre 6,5 e 7,5 e temperatura entre 20 e 30 °C. Esses intervalos são exclusivamente didáticos.

```python
pH = 7.1
T = 26.0

ph_adequado = 6.5 <= pH <= 7.5
t_adequada = 20.0 <= T <= 30.0

if ph_adequado and t_adequada:
    print("Amostra dentro das duas especificações didáticas.")
else:
    print("Ao menos uma especificação não foi atendida.")
```

**Atenção:** `and` não é o mesmo que `or`. Usar `or` no exemplo anterior aprovaria uma amostra que atendesse a apenas uma das duas condições.

<div id="secao-6" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  6. Instrumentação e Controle de Qualidade
</div>

**Exemplo — transmissor linear de pressão.** Para uma faixa de 0 a 10 bar associada a 4–20 mA, adotamos a relação ideal

$$P=\frac{I-4}{16}(10-0)$$

Antes de calcular a pressão, verificamos se a corrente está na faixa nominal. O modelo linear é uma simplificação didática; instrumentos reais requerem calibração e análise de incertezas.

**Cálculo manual:** para $I=12$ mA, temos $P=(12-4)	\times 10/16=5$ bar.

```python
I = 12.0  # mA

if 4.0 <= I <= 20.0:
    P = (I - 4.0) * 10.0 / 16.0
    print(f"Pressão estimada = {P:.2f} bar")
else:
    print("Corrente fora da faixa nominal: cálculo não realizado.")
```

**Interpretação de Engenharia.** A condição impede que a fórmula seja aplicada automaticamente a uma leitura fora da faixa nominal. Isso não significa que o instrumento esteja necessariamente defeituoso.

<div id="secao-7" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  7. Controle de Processos e Controle Aplicado
</div>

**Exemplo — erro de controle.** Considere o erro $e=T_{SP}-T_{PV}$. Para um exemplo didático, vamos classificar o erro usando uma tolerância de 2 °C. A classificação não substitui a lógica de um controlador industrial.

Se $T_{SP}=80$ °C e $T_{PV}=76$ °C, então $e=80-76=4$ °C. O erro positivo indica temperatura medida abaixo da referência.

```python
T_SP = 80.0
T_PV = 76.0
tolerancia = 2.0

erro = T_SP - T_PV

if erro > tolerancia:
    print("Temperatura abaixo da referência, fora da tolerância.")
elif erro < -tolerancia:
    print("Temperatura acima da referência, fora da tolerância.")
else:
    print("Temperatura dentro da tolerância definida.")

print(f"Erro = {erro:.1f} °C")
```

**Controle aplicado:** o sinal do erro, isoladamente, não determina se devemos abrir ou fechar uma válvula. Essa decisão depende da ação do controlador, do atuador e do comportamento físico do processo.

<div id="secao-8" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  8. Mecânica dos Fluidos
</div>

**Exemplo — classificação do regime de escoamento em tubo circular.** O número de Reynolds é

$$Re=\frac{
ho vD}{\mu}$$

com $
ho$ em kg/m³, $v$ em m/s, $D$ em m e $\mu$ em Pa·s. Para escoamento interno em tubo circular, adotaremos a classificação usual aproximada: laminar para $Re<2300$, transição para $2300\leq Re\leq4000$ e turbulento para $Re>4000$. As fronteiras são aproximadas e dependem das condições de escoamento.

**Cálculo manual:** para água com $
ho=1000$ kg/m³, $\mu=0{,}001$ Pa·s, $v=0{,}10$ m/s e $D=0{,}02$ m, resulta $Re=2000$.

```python
rho = 1000.0   # kg/m³
mu = 0.001      # Pa.s
v = 0.10        # m/s
D = 0.02        # m

Re = rho * v * D / mu

if Re < 2300:
    regime = "Laminar"
elif Re <= 4000:
    regime = "Transição"
else:
    regime = "Turbulento"

print(f"Re = {Re:.0f} | Regime: {regime}")
```

**Interpretação de Engenharia.** A escolha de uma correlação de perda de carga depende, entre outros fatores, do regime de escoamento. Uma estrutura condicional pode auxiliar nessa seleção, desde que o modelo e sua faixa de validade sejam conhecidos.

<div id="secao-9" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  9. Transferência de Calor
</div>

**Exemplo — condução de calor em parede plana.** Sob hipóteses de condução unidimensional e regime estacionário, a taxa de calor com sinal pode ser escrita como

$$\dot Q=\frac{kA(T_1-T_2)}{L}$$

Aqui, o sinal indica o sentido adotado para o fluxo de calor. Uma diferença de temperatura negativa não é, por si só, um erro: significa inversão do sentido do fluxo.

**Cálculo manual:** com $k=0{,}8$ W/(m·K), $A=2$ m², $L=0{,}10$ m, $T_1=90$ °C e $T_2=40$ °C, obtemos $\dot Q=800$ W.

```python
k = 0.8       # W/(m.K)
A = 2.0       # m²
L = 0.10      # m
T1 = 90.0     # °C
T2 = 40.0     # °C

if L <= 0 or k <= 0 or A <= 0:
    print("Parâmetros geométricos ou térmicos inválidos.")
else:
    Q = k * A * (T1 - T2) / L
    if Q > 0:
        sentido = "De 1 para 2"
    elif Q < 0:
        sentido = "De 2 para 1"
    else:
        sentido = "Sem transferência líquida por condução"
    print(f"Taxa com sinal = {Q:.1f} W | {sentido}")
```

Este exemplo também introduz um `if` **aninhado**: a classificação do sentido só acontece depois que os parâmetros são considerados válidos.

  10. Modelagem, Simulação e Otimização
</div>

**Exemplo — seleção de um modelo cinético em função da temperatura.** Em Modelagem e Simulação, condições diferentes podem levar à escolha de modelos diferentes. A seleção deve sempre ter justificativa experimental. Aqui utilizamos um critério **fictício**, apenas para ensinar a lógica computacional: abaixo de 330 K usamos um modelo A; entre 330 e 360 K, um modelo B; acima de 360 K, o modelo não está validado.

Isso também mostra uma boa prática: não extrapolar automaticamente um modelo além da faixa de validade declarada.

```python
T = 345.0  # K

if T < 330.0:
    modelo = "Modelo A (faixa didática inferior)"
elif T <= 360.0:
    modelo = "Modelo B (faixa didática superior)"
else:
    modelo = "Fora da faixa de validação didática"

print(modelo)
```

**Otimização:** antes de comparar condições operacionais, podemos usar condicionais para descartar soluções que não atendem às restrições de um problema. Não se deve confundir uma solução matematicamente possível com uma operação industrial segura.

<div id="secao-11" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  11. Processos Químicos e Ciências dos Materiais
</div>

**Processos Químicos de Fabricação — rendimento de reação.** Em um exemplo didático, classificaremos o rendimento de um lote segundo critérios hipotéticos de qualidade: abaixo de 80%, revisão necessária; de 80% até menos de 95%, dentro da faixa intermediária; a partir de 95%, rendimento elevado. Esses limites não representam especificações reais de uma indústria.

```python
rendimento = 92.0  # %

if not 0 <= rendimento <= 100:
    print("Rendimento percentual inválido.")
elif rendimento < 80:
    print("Rendimento abaixo do critério didático.")
elif rendimento < 95:
    print("Rendimento na faixa intermediária didática.")
else:
    print("Rendimento elevado segundo o critério didático.")
```

**Ciências dos Materiais — classificação ilustrativa de dureza.** Uma medição de dureza pode ser comparada a uma faixa de aceitação definida para um material e um método de ensaio específicos. No exemplo, a faixa de 180 a 220 HV é **hipotética**:

```python
dureza = 205.0  # HV (exemplo didático)

if dureza < 180.0:
    print("Abaixo da faixa hipotética de aceitação.")
elif dureza <= 220.0:
    print("Dentro da faixa hipotética de aceitação.")
else:
    print("Acima da faixa hipotética de aceitação.")
```

**Interpretação de Engenharia.** Uma medida de dureza isolada não determina todas as propriedades mecânicas ou a adequação de um material a uma aplicação. É necessário considerar método, carga, microestrutura e especificações técnicas.
