<div id="secao-1" style="background-color: #f1faff; border-left: 6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">1. <strong>Por que organizar informações de processos?</strong></div>

Em uma instalação industrial, uma leitura isolada de temperatura tem utilidade limitada se não estiver associada ao instante, à unidade e ao equipamento. Organizar dados significa preservar essas relações. Por exemplo, uma medição pode conter tempo (min), temperatura (°C) e pressão (kPa). Uma **tupla** é adequada para agrupar uma observação de estrutura fixa; um **dicionário** torna explícito o significado de cada variável por meio de uma chave.

A escolha da estrutura não substitui a análise física: antes de calcular, é necessário conhecer unidades, condições de operação e hipóteses adotadas.

```python
tempo_min = 5
temperatura_c = 32.5
pressao_kpa = 102.4
print(f"Tempo: {tempo_min} min | Temperatura: {temperatura_c:.1f} °C | Pressão: {pressao_kpa:.1f} kPa")
```

<div id="secao-2" style="background-color: #f1faff; border-left: 6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">2. <strong>Tuplas: definição e criação</strong></div>

Uma **tupla** (`tuple`) é uma sequência ordenada e imutável. Normalmente escrevemos seus elementos entre parênteses, separados por vírgulas. Os elementos podem ter tipos distintos; contudo, em aplicações técnicas, convém manter um significado e uma ordem documentados. A tupla vazia é `()`; uma tupla com apenas um elemento exige a vírgula: `(25.0,)`.

```python
medicao = (5, 32.5, 102.4)  # (tempo_min, temperatura_c, pressao_kpa)
coordenadas = (2.0, 4.5)
tupla_unitaria = (25.0,)
print("Medição:", medicao)
print("Tipo:", type(medicao).__name__)
print("Tupla unitária:", tupla_unitaria, "| Tipo:", type(tupla_unitaria).__name__)
```

<div id="secao-3" style="background-color: #f1faff; border-left: 6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">3. <strong>Acesso, fatiamento e operações com tuplas</strong></div>

Os índices começam em zero. Índices negativos acessam os elementos a partir do fim. O fatiamento `[inicio:fim]` inclui o índice inicial e exclui o final. Também podemos utilizar `len()`, `count()` e `index()`. Observe que `index()` procura a primeira ocorrência e gera erro se o valor não existir.

```python
medicao = (5, 32.5, 102.4)
print("Tempo (min):", medicao[0])
print("Temperatura (°C):", medicao[1])
print("Pressão (kPa):", medicao[-1])
print("Primeiros dois valores:", medicao[:2])
print("Quantidade de elementos:", len(medicao))
print("Ocorrências de 5:", medicao.count(5))
```

<div id="secao-4" style="background-color: #f1faff; border-left: 6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">4. <strong>Imutabilidade e desempacotamento</strong></div>

A imutabilidade impede substituir, adicionar ou remover diretamente um elemento de uma tupla. Isso ajuda a representar registros que não devem ser alterados acidentalmente. Não significa, porém, que objetos mutáveis contidos na tupla sejam protegidos contra alterações internas. Para produzir um registro corrigido, crie uma nova tupla. O **desempacotamento** associa cada posição a uma variável, tornando o código mais legível.

```python
registro_original = (5, 32.5, 102.4)
tempo_min, temperatura_c, pressao_kpa = registro_original
print(f"No instante {tempo_min} min: T = {temperatura_c:.1f} °C; P = {pressao_kpa:.1f} kPa")
registro_corrigido = (tempo_min, 32.7, pressao_kpa)
print("Original:", registro_original)
print("Corrigido:", registro_corrigido)
```

<div id="secao-5" style="background-color: #f1faff; border-left: 6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">5. <strong>Tuplas em registros experimentais</strong></div>

Considere quatro leituras realizadas em um tanque. Cada tupla representa uma observação na ordem `(tempo, temperatura, pressão)`. A coleção de observações é uma **lista de tuplas**: a lista permite acrescentar medições, enquanto cada tupla preserva a estrutura de uma leitura. Os dados abaixo são **hipotéticos e didáticos**.

```python
medicoes = [
    (0, 25.0, 101.3),
    (5, 28.0, 101.8),
    (10, 31.5, 102.1),
    (15, 34.0, 102.5),
]
print(f"{'Tempo (min)':>11} | {'T (°C)':>7} | {'P (kPa)':>8}")
print("-" * 34)
for tempo, temperatura, pressao in medicoes:
    print(f"{tempo:11.0f} | {temperatura:7.1f} | {pressao:8.1f}")
```

**Interpretação.** A tabela permite observar a evolução das variáveis ao longo do tempo. O alinhamento com `f-strings` utiliza a sintaxe `{valor:largura.casas_decimaisf}`: a largura é mínima e inclui os caracteres do número, e a precisão indica as casas após o ponto. A associação entre posições e unidades precisa ser mantida em todo o programa.

<div id="secao-6" style="background-color: #f1faff; border-left: 6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">6. <strong>Dicionários: pares chave–valor</strong></div>

Um **dicionário** (`dict`) associa **chaves** a valores. Em vez de lembrar que a temperatura está na posição 1, podemos acessá-la pela chave `'temperatura_c'`. As chaves devem ser únicas e de tipo hashable (por exemplo, strings, números ou tuplas de elementos hashable). Os valores podem ter diferentes tipos. Dicionários preservam a ordem de inserção no Python moderno, mas são usados principalmente para acesso por chave, não por posição.

```python
tanque = {
    "identificacao": "TQ-101",
    "volume_l": 500.0,
    "temperatura_c": 34.0,
    "pressao_kpa": 102.5,
    "em_operacao": True,
}
print("Equipamento:", tanque["identificacao"])
print(f"Temperatura: {tanque['temperatura_c']:.1f} °C")
print("Chaves:", list(tanque.keys()))
```

<div id="secao-7" style="background-color: #f1faff; border-left: 6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">7. <strong>Consulta, atualização e remoção de dados</strong></div>

O acesso com colchetes (`dados[chave]`) exige que a chave exista; caso contrário, ocorre `KeyError`. O método `.get(chave, padrao)` permite informar um valor alternativo. A atribuição atualiza ou cria uma chave; `pop()` remove uma chave e retorna seu valor. Em registros experimentais, atualizações devem ser documentadas para preservar a rastreabilidade.

```python
tanque = {"identificacao": "TQ-101", "temperatura_c": 34.0, "pressao_kpa": 102.5}
print("Vazão cadastrada:", tanque.get("vazao_l_min", "não informada"))
tanque["vazao_l_min"] = 12.0
tanque["temperatura_c"] = 34.2
print("Após atualização:", tanque)
valor_removido = tanque.pop("vazao_l_min")
print("Vazão retirada do registro:", valor_removido, "L/min")
```

<div id="secao-8" style="background-color: #f1faff; border-left: 6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">8. <strong>Percorrendo dicionários e realizando cálculos</strong></div>

O método `.items()` fornece pares `(chave, valor)` para percorrer um dicionário. Quando precisamos calcular médias, máximos ou mínimos, devemos selecionar apenas variáveis numericamente comparáveis, com **mesma grandeza e unidade**. Não faria sentido calcular uma média entre pressão e temperatura apenas porque ambas são números.

```python
temperaturas = {"TQ-101": 34.2, "TQ-102": 31.8, "TQ-103": 36.0}
for equipamento, temperatura in temperaturas.items():
    print(f"{equipamento}: {temperatura:.1f} °C")
media = sum(temperaturas.values()) / len(temperaturas)
mais_quente = max(temperaturas, key=temperaturas.get)
print(f"Temperatura média dos tanques: {media:.2f} °C")
print(f"Maior temperatura: {mais_quente} ({temperaturas[mais_quente]:.1f} °C)")
```

**Atenção à interpretação:** a média calculada é uma média aritmética entre equipamentos. Se os volumes dos tanques forem diferentes, ela não representa necessariamente a temperatura média ponderada de um inventário de fluido.

<div id="secao-9" style="background-color: #f1faff; border-left: 6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">9. <strong>Estruturas combinadas: listas, tuplas e dicionários</strong></div>

Estruturas podem ser combinadas conforme o problema. Uma **lista de dicionários** é útil quando cada linha de uma tabela experimental deve carregar nomes de variáveis explícitos. Um **dicionário de tuplas** pode ser útil para associar um identificador a coordenadas fixas ou faixas de referência. O importante é não aumentar a complexidade sem necessidade.

```python
registros = [
    {"tempo_min": 0, "temperatura_c": 25.0, "pressao_kpa": 101.3},
    {"tempo_min": 5, "temperatura_c": 28.0, "pressao_kpa": 101.8},
    {"tempo_min": 10, "temperatura_c": 31.5, "pressao_kpa": 102.1},
]
for registro in registros:
    print(f"t = {registro['tempo_min']:2d} min | T = {registro['temperatura_c']:4.1f} °C | P = {registro['pressao_kpa']:5.1f} kPa")
limites_didaticos = {"temperatura_c": (20.0, 40.0), "pressao_kpa": (100.0, 105.0)}
print("Faixa didática de temperatura:", limites_didaticos["temperatura_c"], "°C")
```

<div id="secao-10" style="background-color: #f1faff; border-left: 6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">10. <strong>Estudo de caso: monitoramento de um tanque</strong></div>

**Problema de Engenharia.** Um tanque é acompanhado em intervalos de cinco minutos. Deseja-se organizar os registros, identificar temperaturas fora de uma faixa de referência **hipotética** e estimar a taxa média de variação da temperatura entre a primeira e a última medição.

**Modelo matemático.** Para instantes inicial e final distintos, a taxa média é

$$\overline{r}_T = \frac{T_f-T_i}{t_f-t_i}$$

em °C/min. Trata-se da variação média observada no intervalo, não de uma derivada instantânea nem de uma previsão da evolução futura. Adotaremos, apenas para o exercício, uma faixa de 20 a 35 °C. Ela **não é um limite de segurança industrial**.

```python
processo = {
    "equipamento": "TQ-101",
    "unidades": {"tempo": "min", "temperatura": "°C", "pressao": "kPa"},
    "faixa_temperatura_c": (20.0, 35.0),
    "medicoes": [
        {"tempo_min": 0, "temperatura_c": 25.0, "pressao_kpa": 101.3},
        {"tempo_min": 5, "temperatura_c": 28.0, "pressao_kpa": 101.8},
        {"tempo_min": 10, "temperatura_c": 31.5, "pressao_kpa": 102.1},
        {"tempo_min": 15, "temperatura_c": 36.0, "pressao_kpa": 102.6},
    ],
}
limite_inferior, limite_superior = processo["faixa_temperatura_c"]
leituras = processo["medicoes"]
print(f"Monitoramento do equipamento {processo['equipamento']}")
print(f"{'t (min)':>7} | {'T (°C)':>7} | {'P (kPa)':>8} | Situação")
print("-" * 49)
for leitura in leituras:
    t = leitura["tempo_min"]
    T = leitura["temperatura_c"]
    P = leitura["pressao_kpa"]
    situacao = "na faixa" if limite_inferior <= T <= limite_superior else "fora da faixa"
    print(f"{t:7.0f} | {T:7.1f} | {P:8.1f} | {situacao}")

primeira, ultima = leituras[0], leituras[-1]
delta_t = ultima["tempo_min"] - primeira["tempo_min"]
if delta_t <= 0:
    raise ValueError("Os instantes final e inicial devem ser distintos e crescentes.")
taxa_media = (ultima["temperatura_c"] - primeira["temperatura_c"]) / delta_t
print(f"\nTaxa média de variação da temperatura: {taxa_media:.3f} °C/min")
```

**Análise dos resultados.** A temperatura aumentou de 25,0 para 36,0 °C em 15 minutos. Assim, a taxa média foi `(36,0 − 25,0)/15 = 0,733 °C/min` (aproximadamente). A última leitura está acima da faixa didática adotada. Isso demonstra como a estruturação dos registros permite transformar dados em informação útil, mas a decisão operacional real exigiria limites validados, conhecimento do equipamento e procedimentos de segurança.
