1. Por que utilizar listas na Engenharia
</div>

Considere um transmissor que registra a pressão de uma linha em cinco instantes. Se cada medição for armazenada em uma variável diferente, o programa ficará difícil de ampliar e de verificar. Uma **lista** permite reunir os valores na ordem de aquisição e aplicar a mesma operação a todos eles. A lista não substitui a interpretação física: é necessário conhecer as unidades, o significado dos dados e a ordem em que foram obtidos.

````python
# Pressões medidas em uma linha de processo (kPa)
pressoes_kpa = [198.0, 201.5, 199.8, 203.2, 200.5]
print("Medições (kPa):", pressoes_kpa)
print("Número de medições:", len(pressoes_kpa))
````

**Interpretação.** A ordem dos valores pode representar a sequência temporal das medições. `len()` informa quantas observações foram registradas; não informa, por si só, a duração do ensaio.

2. Criando listas e reconhecendo seus elementos
</div>

Uma lista é escrita entre colchetes, com elementos separados por vírgulas. Ela pode armazenar números, textos e outros objetos. Em aplicações de Engenharia, é recomendável manter dados de uma mesma grandeza em uma lista e registrar a unidade no nome da variável ou na documentação.

````python
temperaturas_c = [28.5, 29.1, 30.0, 30.8]
equipamentos = ["Tanque T-101", "Bomba P-101", "Trocador E-101"]
lista_vazia = []
print("Temperaturas:", temperaturas_c)
print("Equipamentos:", equipamentos)
print("Tipo da estrutura:", type(temperaturas_c).__name__)
print("Quantidade de temperaturas:", len(temperaturas_c))
````

**Atenção:** uma lista pode conter valores de tipos diferentes, mas misturar temperaturas, pressões e vazões na mesma sequência geralmente dificulta os cálculos e aumenta o risco de erros de unidade.

3. Índices, acesso e fatiamento
</div>

Os índices começam em **zero**: o primeiro elemento ocupa a posição `0`, o segundo a posição `1` e assim por diante. Índices negativos contam a partir do final. O fatiamento `lista[início:fim]` inclui o índice inicial e exclui o final.

````python
temperaturas_c = [28.5, 29.1, 30.0, 30.8, 31.4, 31.9]
print("Primeira medição:", temperaturas_c[0], "°C")
print("Última medição:", temperaturas_c[-1], "°C")
print("Três primeiras:", temperaturas_c[:3])
print("Da terceira à quinta:", temperaturas_c[2:5])
print("Medições alternadas:", temperaturas_c[::2])
````

**Verificação:** tentar acessar `temperaturas_c[6]` provoca `IndexError`, pois os índices válidos desta lista vão de `0` a `5`. Para evitar erros, verifique o comprimento com `len()` antes de acessar posições variáveis.

4. Alterando e organizando listas
</div>

Listas são **mutáveis**: seus elementos podem ser alterados após a criação. Isso é útil para corrigir uma leitura identificada como incorreta ou acrescentar novos registros. Contudo, modificar uma lista de medições originais pode prejudicar a rastreabilidade; em trabalhos experimentais, preserve os dados brutos e faça correções em uma cópia.

````python
leituras_originais = [99.8, 100.1, 999.0, 100.3]
leituras_corrigidas = leituras_originais.copy()
leituras_corrigidas[2] = 100.2  # Correção hipotética documentada
leituras_corrigidas.append(100.4)
print("Originais:", leituras_originais)
print("Corrigidas:", leituras_corrigidas)
print("Quantidade de leituras corrigidas:", len(leituras_corrigidas))
````

Os métodos `append()` e `extend()` acrescentam, respectivamente, um elemento e os elementos de outra sequência; `insert()` insere em uma posição; `pop()` remove e retorna um elemento. `sort()` ordena **a própria lista**, enquanto `sorted()` devolve uma nova lista ordenada. **Cuidado:** ordenar apenas os valores medidos destrói sua associação com os instantes de aquisição.

````python
amostras = [3.1, 2.8, 3.4]
ordenadas = sorted(amostras)
print("Ordem de aquisição:", amostras)
print("Ordem crescente:", ordenadas)
````

5. Percorrendo listas com laços
</div>

Para processar várias medições, utilizamos laços. O `for` percorre os elementos sem exigir o controle manual de um índice; `enumerate()` fornece simultaneamente a posição e o valor. Essa abordagem é especialmente útil quando precisamos identificar o instante de uma ocorrência.

````python
pressoes_kpa = [198.0, 201.5, 199.8, 203.2, 200.5]
for indice, pressao in enumerate(pressoes_kpa):
    print(f"Medição {indice + 1}: {pressao:.1f} kPa")

limite_kpa = 202.0
for indice, pressao in enumerate(pressoes_kpa):
    if pressao > limite_kpa:
        print(f"Atenção: medição {indice + 1} acima de {limite_kpa:.1f} kPa")
````

**Interpretação.** O limite adotado é apenas ilustrativo, não um limite operacional certificado. Em sistemas reais, alarmes dependem de critérios de projeto, incerteza de medição e procedimentos de segurança.

6. Cálculos de Engenharia com listas
</div>

Uma lista permite calcular indicadores de um conjunto de medições. Vamos analisar a velocidade média de escoamento de água em uma tubulação para diferentes vazões volumétricas, utilizando a relação física $v=Q/A$, com $A=\pi D^2/4$. É indispensável converter a vazão para $\mathrm{m^3/s}$ antes do cálculo.

**Problema de Engenharia.** Uma tubulação de diâmetro interno $D=0{,}050\,\mathrm{m}$ conduz água com vazões de $6$, $9$ e $12\,\mathrm{m^3/h}$. Determine as velocidades médias correspondentes.

A área transversal é

$$
A=\frac{\pi D^2}{4}=\frac{\pi(0{,}050)^2}{4}\approx 1{,}9635\times10^{-3}\,\mathrm{m^2}.
$$

Para $Q=6\,\mathrm{m^3/h}$, a conversão fornece $Q=6/3600\,\mathrm{m^3/s}$, de modo que $v\approx0{,}849\,\mathrm{m/s}$. Vamos repetir o cálculo para todos os valores.

````python
from math import pi

vazoes_m3h = [6.0, 9.0, 12.0]
diametro_m = 0.050
area_m2 = pi * diametro_m**2 / 4
velocidades_ms = []
for vazao_m3h in vazoes_m3h:
    vazao_m3s = vazao_m3h / 3600
    velocidade_ms = vazao_m3s / area_m2
    velocidades_ms.append(velocidade_ms)

for vazao, velocidade in zip(vazoes_m3h, velocidades_ms):
    print(f"Q = {vazao:4.1f} m³/h -> v = {velocidade:.3f} m/s")
````

**Interpretação física.** Como o diâmetro é constante, a área também é constante; portanto, a velocidade cresce linearmente com a vazão. Os resultados são velocidades **médias na seção transversal**, não perfis locais de velocidade.

Outra tarefa frequente é resumir medições. A média aritmética é $\bar{x}=\frac{1}{n}\sum_{i=1}^{n}x_i$. O intervalo entre máximo e mínimo mostra a amplitude observada, mas não substitui uma análise de incerteza.

````python
temperaturas_c = [72.0, 72.4, 73.1, 72.8, 73.3]
media_c = sum(temperaturas_c) / len(temperaturas_c)
amplitude_c = max(temperaturas_c) - min(temperaturas_c)
print(f"Temperatura média: {media_c:.2f} °C")
print(f"Amplitude observada: {amplitude_c:.2f} °C")
````

7. Listas e funções
</div>

Funções permitem encapsular cálculos que serão reutilizados com diferentes conjuntos de dados. Em vez de depender de uma lista global, a função deve receber as medições como argumento e devolver um resultado. Essa separação facilita testes, revisão e reutilização.

````python
def media_medicoes(valores):
    """Retorna a média aritmética de uma sequência numérica não vazia."""
    if len(valores) == 0:
        raise ValueError("A lista de medições não pode estar vazia.")
    return sum(valores) / len(valores)

pressao_a_kpa = [200.0, 201.0, 199.0]
pressao_b_kpa = [205.0, 204.0, 206.0]
print(f"Média A: {media_medicoes(pressao_a_kpa):.2f} kPa")
print(f"Média B: {media_medicoes(pressao_b_kpa):.2f} kPa")
````

**Discussão.** A validação da lista vazia evita uma divisão por zero. O nome e a documentação da função deixam claro que se trata de uma média aritmética simples, adequada quando cada medição deve ter o mesmo peso.

8. Listas aninhadas e tabelas de medições
</div>

Quando há várias variáveis medidas nos mesmos instantes, podemos utilizar listas aninhadas: cada lista interna representa uma linha de dados. Para manter a interpretação, definiremos explicitamente a ordem das colunas: tempo, temperatura e pressão. Em aplicações maiores, estruturas tabulares especializadas podem ser mais convenientes; aqui o objetivo é compreender a organização básica.

````python
# Colunas: [tempo (min), temperatura (°C), pressão (kPa)]
medicoes = [
    [0, 65.0, 200.0],
    [2, 66.2, 201.5],
    [4, 67.1, 201.0],
    [6, 68.0, 202.2],
]
print("Tempo (min) | T (°C) | P (kPa)")
for linha in medicoes:
    tempo_min, temperatura_c, pressao_kpa = linha
    print(f"{tempo_min:11.0f} | {temperatura_c:6.1f} | {pressao_kpa:7.1f}")
````

**Ponto importante.** `medicoes[1][2]` acessa a pressão da segunda linha. Em tabelas desse tipo, cada linha deve manter o mesmo número de colunas e as mesmas unidades. Uma lista aninhada não identifica automaticamente o significado físico dos campos.

9. Compreensões de listas
</div>

Uma compreensão de listas é uma forma compacta de construir uma nova lista a partir de uma sequência. Ela pode tornar uma transformação simples mais legível, mas não deve esconder operações físicas complexas. Começaremos convertendo temperaturas de graus Celsius para kelvin pela relação $T_K=T_{°C}+273{,}15$.


__Entendendo a formatação numérica com `f-strings`__

Ao apresentar resultados de cálculos ou medições experimentais, é importante organizar os valores numéricos em colunas, facilitando a leitura e a comparação dos dados. Em Python, podemos utilizar as chamadas **f-strings** para controlar tanto a quantidade de casas decimais quanto o espaço reservado para cada número.

Considere o seguinte comando:

```python
print(f"{tempo_min:11.0f} | {temperatura_c:6.1f} | {pressao_kpa:7.1f}")
```

A estrutura geral da formatação é:

`{variavel:largura.casas_decimaisf}`

O número **antes do ponto** representa a largura mínima reservada para apresentar o valor, incluindo os algarismos, o ponto decimal, o sinal negativo (quando existir) e os espaços necessários. O número **depois do ponto** determina a quantidade de casas decimais. A letra `f` indica que o número será apresentado em notação decimal de ponto fixo.

No exemplo apresentado:

- **`11.0f`** reserva uma largura mínima de 11 caracteres e apresenta o valor sem casas decimais.
- **`6.1f`** reserva uma largura mínima de 6 caracteres e apresenta o valor com uma casa decimal.
- **`7.1f`** reserva uma largura mínima de 7 caracteres e apresenta o valor com uma casa decimal.

Por padrão, os valores numéricos são alinhados à direita dentro do espaço reservado. Quando o número ocupa menos caracteres que a largura especificada, o Python acrescenta espaços à esquerda para completar essa largura.

Por exemplo, o valor `28.5`, formatado com `6.1f`, ocupa quatro caracteres. Como foram reservados seis, o Python acrescenta dois espaços à esquerda.

**É importante observar que a largura especificada representa um valor mínimo, e não um limite máximo.** Caso o número necessite de mais caracteres, o Python utilizará o espaço necessário para apresentá-lo integralmente.

Esse recurso é especialmente útil na Engenharia, pois permite construir tabelas organizadas com valores de temperatura, pressão, vazão, tempo e outras variáveis de processo, mantendo o alinhamento das colunas e facilitando a interpretação dos resultados.


````python
temperaturas_c = [20.0, 25.0, 30.0, 35.0]
temperaturas_k = [temperatura + 273.15 for temperatura in temperaturas_c]
print("°C:", temperaturas_c)
print("K :", temperaturas_k)

# Seleção de temperaturas estritamente acima de 28 °C
acima_28_c = [temperatura for temperatura in temperaturas_c if temperatura > 28]
print("Acima de 28 °C:", acima_28_c)
````

**Comparação.** O `for` tradicional é conveniente quando o procedimento exige várias etapas e comentários; a compreensão é apropriada para transformações diretas e curtas. Ambas as formas devem produzir os mesmos valores.

10. Estudo de caso: monitoramento de um processo
</div>

Vamos integrar os conceitos em um pequeno estudo de monitoramento de temperatura de um tanque. Foram realizadas seis medições em intervalos regulares de dois minutos. Nosso objetivo é organizar as leituras, calcular indicadores, identificar valores acima de um limite ilustrativo e estimar a taxa média de variação da temperatura no intervalo completo.

**Dados do problema.** Considere os tempos $t=[0,2,4,6,8,10]\,\mathrm{min}$ e as temperaturas $T=[60{,}0;61{,}2;62{,}1;63{,}5;64{,}0;65{,}3]\,°\mathrm{C}$. Para estimar a taxa média de aquecimento, utilizamos


$$ 
\frac{\Delta T}{\Delta t}=\frac{T_f-T_i}{t_f-t_i} 
$$


Esse quociente descreve a variação média no intervalo e não representa necessariamente a derivada instantânea da temperatura.

````python
tempos_min = [0, 2, 4, 6, 8, 10]
temperaturas_c = [60.0, 61.2, 62.1, 63.5, 64.0, 65.3]
limite_c = 64.0  # Limite didático, não especificação de segurança

if len(tempos_min) != len(temperaturas_c):
    raise ValueError("Tempo e temperatura devem ter o mesmo número de valores.")
if len(tempos_min) < 2 or tempos_min[-1] <= tempos_min[0]:
    raise ValueError("São necessários ao menos dois tempos com intervalo positivo.")

media_c = sum(temperaturas_c) / len(temperaturas_c)
taxa_media_c_min = ((temperaturas_c[-1] - temperaturas_c[0]) /
                    (tempos_min[-1] - tempos_min[0]))
acima_limite = [(t, T) for t, T in zip(tempos_min, temperaturas_c)
                if T > limite_c]

print("MONITORAMENTO DO TANQUE")
for t, T in zip(tempos_min, temperaturas_c):
    print(f"t = {t:2d} min | T = {T:4.1f} °C")
print(f"Temperatura média: {media_c:.2f} °C")
print(f"Taxa média de aquecimento: {taxa_media_c_min:.2f} °C/min")
print("Medições acima do limite didático:", acima_limite)
````

**Análise dos resultados.** A temperatura aumenta de $60{,}0$ para $65{,}3\,°\mathrm{C}$ em dez minutos, produzindo taxa média de $0{,}53\,°\mathrm{C/min}$. O valor de $64{,}0\,°\mathrm{C}$ não entra na seleção porque a condição adotada foi **estritamente maior** (`>`), e não maior ou igual (`>=`). Uma sequência crescente, por si só, não permite concluir qual mecanismo físico está causando o aquecimento.
