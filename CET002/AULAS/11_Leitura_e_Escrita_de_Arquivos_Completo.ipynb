Uma lista Python guarda valores enquanto o programa está em execução. Um arquivo preserva esses valores para consultas posteriores. É importante registrar também **unidades**, **instantes de medição** e **identificação do ensaio**: o número `80`, isoladamente, não indica se estamos falando de temperatura, pressão ou concentração.

Os arquivos que criaremos serão didáticos e não representam registros industriais certificados.

<div id="secao-2" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:20px; font-weight:bold;">2. Pastas e caminhos</div>

Utilizaremos a biblioteca padrão `pathlib` para criar uma pasta de trabalho. Ela evita depender de separadores específicos de Windows ou Linux. Observe que a pasta de execução do Jupyter pode ser diferente da pasta onde o notebook foi salvo.

```python
from pathlib import Path
pasta = Path("dados_notebook11")
pasta.mkdir(exist_ok=True)
print("Diretório de execução:", Path.cwd())
print("Diretório de dados:", pasta.resolve())
```

**Interpretação:** os exemplos seguintes criarão arquivos dentro de `dados_notebook11`, sem alterar outros documentos do projeto.

<div id="secao-3" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:20px; font-weight:bold;">3. Gravando medições em texto</div>

**Transferência de Calor — registro de temperaturas.** Considere quatro leituras sucessivas: 65, 68, 72 e 75 °C. Desejamos armazená-las em um arquivo `.txt`. O modo `"w"` cria o arquivo, mas **substitui seu conteúdo se ele já existir**. O comando `with` garante o fechamento correto do arquivo.

```python
temperaturas = [65, 68, 72, 75]
arquivo_txt = pasta / "temperaturas.txt"
with open(arquivo_txt, "w", encoding="utf-8") as arquivo:
    arquivo.write("Temperatura (°C)\n")
    for temperatura in temperaturas:
        arquivo.write(f"{temperatura}\n")
print("Arquivo criado:", arquivo_txt)
```

**Interpretação:** cada temperatura ocupa uma linha. O caractere `\n` indica a quebra de linha.

<div id="secao-4" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:20px; font-weight:bold;">4. Lendo e analisando medições</div>

Para recuperar as temperaturas, usaremos o modo `"r"`. O primeiro registro é o cabeçalho e não deve ser convertido para número. A média das medições será comparada ao cálculo manual:

$$
\overline{T}=\frac{65+68+72+75}{4}=70\ \mathrm{°C}.
$$

```python
with open(arquivo_txt, "r", encoding="utf-8") as arquivo:
    cabecalho = arquivo.readline().strip()
    valores = [float(linha.strip()) for linha in arquivo if linha.strip()]
media = sum(valores) / len(valores)
print("Grandeza:", cabecalho)
print("Valores:", valores)
print(f"Temperatura média: {media:.2f} °C")
```

**Interpretação:** o valor calculado deve ser 70 °C. O método `strip()` remove espaços e quebras de linha; `float()` converte texto numérico para número.

<div id="secao-5" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:20px; font-weight:bold;">5. CSV e registros de instrumentos</div>

**Instrumentação — histórico de pressão.** Um transmissor registra pressão em diferentes instantes. Para dados tabulares, o formato CSV é mais conveniente que um texto sem estrutura. Usaremos `csv.DictWriter`, que escreve os nomes das colunas e as linhas de medição. Os valores são hipotéticos.

```python
import csv
medicoes = [
    {"tempo_min": 0, "pressao_bar": 2.1},
    {"tempo_min": 1, "pressao_bar": 2.4},
    {"tempo_min": 2, "pressao_bar": 2.7},
    {"tempo_min": 3, "pressao_bar": 2.5},
]
arquivo_csv = pasta / "pressao.csv"
with open(arquivo_csv, "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.DictWriter(arquivo, fieldnames=["tempo_min", "pressao_bar"])
    escritor.writeheader()
    escritor.writerows(medicoes)
print("Arquivo CSV criado:", arquivo_csv)
```

**Interpretação:** cada linha representa um instante. O argumento `newline=""` evita linhas em branco extras em alguns sistemas.

<div id="secao-6" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:20px; font-weight:bold;">6. Leitura de CSV e conversão numérica</div>

Na leitura de um CSV, os campos são inicialmente **textos**. Para calcular a pressão média, devemos converter a coluna `pressao_bar` para `float`. O cálculo manual fornece:

$$
\overline{P}=\frac{2{,}1+2{,}4+2{,}7+2{,}5}{4}=2{,}425\ \mathrm{bar}.
$$

```python
with open(arquivo_csv, "r", newline="", encoding="utf-8") as arquivo:
    dados = list(csv.DictReader(arquivo))
pressoes = [float(linha["pressao_bar"]) for linha in dados]
print(f"Pressão média: {sum(pressoes)/len(pressoes):.3f} bar")
print(f"Pressão máxima: {max(pressoes):.3f} bar")
```

**Interpretação:** estatísticas descritivas não devem ser confundidas com avaliação de segurança operacional: esta depende dos limites de projeto e das condições reais do equipamento.

<div id="secao-7" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:20px; font-weight:bold;">7. Acrescentando dados ao histórico</div>

O modo `"a"` acrescenta linhas ao final de um arquivo sem apagar o histórico. Entretanto, executar a mesma célula duas vezes pode duplicar registros. Neste exemplo, verificaremos se o instante já está registrado antes de acrescentá-lo.

```python
novo = {"tempo_min": 4, "pressao_bar": 2.6}
with open(arquivo_csv, "r", newline="", encoding="utf-8") as arquivo:
    instantes = {int(linha["tempo_min"]) for linha in csv.DictReader(arquivo)}
if novo["tempo_min"] not in instantes:
    with open(arquivo_csv, "a", newline="", encoding="utf-8") as arquivo:
        csv.DictWriter(arquivo, fieldnames=["tempo_min", "pressao_bar"]).writerow(novo)
    print("Medição acrescentada.")
else:
    print("Instante já existente; nenhuma duplicação realizada.")
```

**Interpretação:** o arquivo mantém as medições anteriores. Em um sistema industrial, seria conveniente usar uma identificação composta por data, hora e equipamento.

<div id="secao-8" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:20px; font-weight:bold;">8. Controle de Qualidade</div>

**Controle de Qualidade — classificação de amostras.** Um laboratório mede a concentração de quatro amostras. Para este exercício, a faixa hipotética de aceitação é de 9,5 a 10,5 g/L, inclusive. A condição lógica é `9.5 <= concentracao <= 10.5`. Esses limites são apenas didáticos, não especificações de um produto real.

```python
amostras = [("A01", 9.8), ("A02", 10.3), ("A03", 10.7), ("A04", 9.4)]
arquivo_qualidade = pasta / "qualidade.csv"
with open(arquivo_qualidade, "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.writer(arquivo)
    escritor.writerow(["amostra", "concentracao_g_L", "situacao"])
    for nome, concentracao in amostras:
        situacao = "Conforme" if 9.5 <= concentracao <= 10.5 else "Não conforme"
        escritor.writerow([nome, concentracao, situacao])
with open(arquivo_qualidade, "r", newline="", encoding="utf-8") as arquivo:
    resultados = list(csv.DictReader(arquivo))
print("Amostras conformes:", sum(r["situacao"] == "Conforme" for r in resultados))
print("Total de amostras:", len(resultados))
```

**Interpretação:** duas das quatro amostras atendem ao critério didático. O CSV conserva tanto a medição quanto a classificação.

<div id="secao-9" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:20px; font-weight:bold;">9. Controle de Processos Aplicado</div>

**Controle de Processos — histórico de erros.** Considere um sistema com referência $T_{SP}=80$ °C e medições de 74, 76, 78, 79 e 80 °C. O erro em cada instante é definido por

$$
e_k=T_{SP}-T_k.
$$

No primeiro instante, $e_0=80-74=6$ °C. Vamos registrar a evolução em CSV, preparando o terreno para análises futuras de desempenho de controladores.

```python
referencia = 80.0
medidas = [74, 76, 78, 79, 80]
arquivo_controle = pasta / "erros_controle.csv"
with open(arquivo_controle, "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.writer(arquivo)
    escritor.writerow(["instante", "temperatura_C", "erro_C"])
    for instante, temperatura in enumerate(medidas):
        escritor.writerow([instante, temperatura, referencia - temperatura])
print("Erros calculados:", [referencia - t for t in medidas])
```

**Interpretação:** o erro diminui nesses cinco instantes. Isso não basta, isoladamente, para concluir que um sistema de controle é estável.

<div id="secao-10" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:20px; font-weight:bold;">10. Processos Químicos de Fabricação</div>

**Processos Químicos de Fabricação — registros de bateladas.** Diferentes bateladas podem ser processadas em condições distintas. Vamos armazenar temperatura e tempo de três bateladas hipotéticas. Um registro industrial real também exigiria data, produto, equipamento, procedimento e responsável.

```python
bateladas = [
    {"batelada": "B01", "temperatura_C": 180, "tempo_min": 30},
    {"batelada": "B02", "temperatura_C": 190, "tempo_min": 25},
    {"batelada": "B03", "temperatura_C": 185, "tempo_min": 35},
]
arquivo_bateladas = pasta / "bateladas.csv"
with open(arquivo_bateladas, "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.DictWriter(arquivo, fieldnames=list(bateladas[0]))
    escritor.writeheader()
    escritor.writerows(bateladas)
print("Bateladas registradas:", len(bateladas))
```

**Interpretação:** um arquivo tabular ajuda a comparar condições de processamento, mas não substitui procedimentos de qualidade e rastreabilidade.

<div id="secao-11" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:20px; font-weight:bold;">11. Ciências dos Materiais</div>

**Ciências dos Materiais — resultados de dureza.** Considere quatro corpos de prova com durezas hipotéticas de 210, 215, 208 e 217 HV. A média é

$$
\overline{H}=\frac{210+215+208+217}{4}=212{,}5\ \mathrm{HV}.
$$

Vamos salvar cada resultado com a identificação do corpo de prova.

```python
durezas = [210, 215, 208, 217]
arquivo_dureza = pasta / "dureza.csv"
with open(arquivo_dureza, "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.writer(arquivo)
    escritor.writerow(["corpo_de_prova", "dureza_HV"])
    for numero, dureza in enumerate(durezas, start=1):
        escritor.writerow([numero, dureza])
with open(arquivo_dureza, "r", newline="", encoding="utf-8") as arquivo:
    valores = [float(r["dureza_HV"]) for r in csv.DictReader(arquivo)]
print(f"Dureza média: {sum(valores)/len(valores):.2f} HV")
```

**Interpretação:** o resultado médio não caracteriza sozinho a dispersão nem a qualidade do material; é necessário conhecer as condições e o método de ensaio.

<div id="secao-12" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:20px; font-weight:bold;">12. JSON e parâmetros de modelos</div>

**Modelagem e Simulação — parâmetros de um tanque.** O JSON permite guardar parâmetros identificados por nomes, facilitando a reprodução de simulações. Para um tanque com $V_0=50$ L, $Q_{in}=5$ L/min e $Q_{out}=4$ L/min, o balanço simplificado fornece

$$
\frac{dV}{dt}=Q_{in}-Q_{out},
$$

$$
V(t)=V_0+(Q_{in}-Q_{out})t.
$$

Após 10 min, $V(10)=50+(5-4)\times10=60$ L. O modelo considera vazões constantes e ausência de outras entradas ou saídas.

```python
import json
parametros = {
    "equipamento": "tanque_didatico",
    "volume_inicial_L": 50.0,
    "vazao_entrada_L_min": 5.0,
    "vazao_saida_L_min": 4.0,
    "tempo_min": 10.0,
}
arquivo_json = pasta / "parametros_tanque.json"
with open(arquivo_json, "w", encoding="utf-8") as arquivo:
    json.dump(parametros, arquivo, indent=4, ensure_ascii=False)
with open(arquivo_json, "r", encoding="utf-8") as arquivo:
    dados_tanque = json.load(arquivo)
volume = dados_tanque["volume_inicial_L"] + (
    dados_tanque["vazao_entrada_L_min"] - dados_tanque["vazao_saida_L_min"]
) * dados_tanque["tempo_min"]
print(f"Volume calculado: {volume:.1f} L")
```

**Interpretação:** o programa recupera os parâmetros de um arquivo externo ao código e calcula o resultado do modelo. Isso facilita a repetição da simulação com outros valores.

<div id="secao-13" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:20px; font-weight:bold;">13. Erros comuns e confiabilidade</div>

Arquivos podem estar ausentes, conter cabeçalhos incorretos ou valores que não podem ser convertidos para números. Um algoritmo confiável deve reconhecer esses problemas antes de apresentar resultados. Vamos tratar os casos de arquivo inexistente e dados inválidos com `try` e `except`.

```python
def ler_temperaturas(caminho):
    try:
        with open(caminho, "r", encoding="utf-8") as arquivo:
            cabecalho = arquivo.readline().strip()
            if cabecalho != "Temperatura (°C)":
                raise ValueError("Cabeçalho inesperado")
            valores = [float(linha.strip()) for linha in arquivo if linha.strip()]
        if not valores:
            raise ValueError("Arquivo sem medições")
        return valores
    except FileNotFoundError:
        print("Erro: arquivo não encontrado")
    except ValueError as erro:
        print("Erro nos dados:", erro)
    return None

print("Leitura válida:", ler_temperaturas(arquivo_txt))
print("Leitura ausente:", ler_temperaturas(pasta / "inexistente.txt"))
```

**Boas práticas:** registrar unidades e identificação do equipamento; conferir cabeçalhos; converter números explicitamente; evitar sobrescrever arquivos importantes; não executar códigos recebidos de fontes desconhecidas; e executar as células deste notebook de cima para baixo.

**Observação sobre planilhas brasileiras:** alguns CSV utilizam ponto e vírgula como separador e vírgula decimal. Nesse caso, configure `delimiter=";"` e trate a conversão decimal de maneira explícita. O formato dos dados deve ser conhecido antes da leitura.
