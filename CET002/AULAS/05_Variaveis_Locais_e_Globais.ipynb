  1. Introdução ao conceito de escopo de variáveis
</div>

Em Engenharia, uma variável pode representar temperatura, pressão, vazão ou um parâmetro de um modelo. Ao organizar os cálculos em funções, precisamos saber **onde cada variável pode ser consultada**. Esse conjunto de regiões é chamado de *escopo*. Uma variável criada dentro de uma função normalmente é local; uma variável definida no nível principal do notebook é global naquele módulo.

**Pergunta orientadora:** se uma função calcula o erro de temperatura, será que outra parte do programa consegue acessar diretamente a variável `erro` criada dentro dela? Vamos investigar executando exemplos pequenos.

<div id="secao-2" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  2. O que é uma variável local?
</div>

Uma variável **local** nasce durante a execução da função e é utilizada no seu corpo. Considere o erro de temperatura de um processo, definido por $e=T_{SP}-T_{PV}$. Para $T_{SP}=80$ °C e $T_{PV}=76$ °C, obtemos $e=4$ °C.

```python
def calcular_erro(T_SP, T_PV):
    erro = T_SP - T_PV  # variável local
    return erro

resultado = calcular_erro(80, 76)
print("Erro de temperatura:", resultado, "°C")
```

A função devolve o valor por meio de `return`, mas a variável local `erro` não passa a existir no escopo principal. Experimente executar a próxima célula: tratamos o erro de propósito para que o notebook continue funcionando.

```python
try:
    print(erro)
except NameError:
    print("A variável local erro não está disponível fora da função.")
```

<div id="secao-3" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  3. O que é uma variável global?
</div>

Uma variável definida fora das funções, no nível principal do notebook, pertence ao **escopo global do módulo**. Uma função pode consultar seu valor, desde que esse nome não seja substituído por uma variável local. No exemplo, `Kc` representa o ganho de um controlador proporcional em pontos percentuais por °C.

```python
Kc = 2.0  # ganho definido fora da função

def acao_proporcional(erro):
    return Kc * erro

print("Contribuição proporcional:", acao_proporcional(4), "%")
```

O cálculo funciona, mas depende de um valor definido fora da função. Se `Kc` mudar em outra célula do notebook, o resultado também poderá mudar sem que a chamada da função deixe isso evidente.

<div id="secao-4" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  4. Diferenças entre variáveis locais e globais
</div>

Compare os dois escopos. Os **parâmetros** também são nomes locais da função; já uma variável global pode ser lida dentro dela. Uma atribuição comum a um nome dentro da função cria um nome local, a menos que haja uma declaração apropriada.

```python
temperatura_ambiente = 25  # global

def avaliar_temperatura(T_medida):
    diferenca = T_medida - temperatura_ambiente  # diferenca é local
    return diferenca

print("Diferença de temperatura:", avaliar_temperatura(40), "°C")
```

**Interpretação:** `temperatura_ambiente` é consultada pela função; `T_medida` e `diferenca` são locais. Uma função mais independente receberia também a temperatura ambiente como parâmetro.

<div id="secao-5" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  5. Variáveis com o mesmo nome: sombreamento
</div>

Se uma variável local possui o mesmo nome de uma global, o nome local **sombreia** o global dentro daquela função. Isso não significa que a variável global foi modificada.

```python
pressao = 200  # kPa, global

def exemplo_pressao():
    pressao = 150  # kPa, local
    print("Dentro da função:", pressao, "kPa")

exemplo_pressao()
print("Fora da função:", pressao, "kPa")
```

O exemplo apresenta 150 kPa dentro da função e 200 kPa fora dela. Para evitar confusão, convém escolher nomes claros e explicitar as entradas do cálculo.

<div id="secao-6" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  6. A palavra-chave global
</div>

A palavra-chave `global` permite que uma função **atribua um novo valor** a um nome do escopo global. Ela existe na linguagem, mas seu uso deve ser criterioso, pois torna o comportamento do programa dependente de alterações externas.

```python
contador_leituras = 0

def registrar_leitura():
    global contador_leituras
    contador_leituras += 1

registrar_leitura()
registrar_leitura()
print("Leituras registradas:", contador_leituras)
```

**Atenção:** consultar uma variável global não exige `global`. A declaração é necessária quando queremos atribuir ao nome global dentro da função. Objetos mutáveis, como listas, exigem uma distinção adicional: modificar o conteúdo de uma lista global não é o mesmo que atribuir uma nova lista ao seu nome.

```python
leituras = [20.0, 21.0]

def adicionar_leitura(valor):
    leituras.append(valor)  # altera a lista existente

adicionar_leitura(22.0)
print("Leituras:", leituras)
```

<div id="secao-7" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  7. Passagem de parâmetros como alternativa às variáveis globais
</div>

Na maior parte dos modelos de Engenharia, é preferível **passar as grandezas como argumentos** e receber o resultado com `return`. Assim, cada chamada deixa explícitos os dados usados. Para o controlador proporcional ideal, $u_P=K_c e$. Se $K_c=2$ %/°C e $e=4$ °C, a contribuição é de 8 pontos percentuais.

```python
def controlador_proporcional(erro, Kc):
    return Kc * erro

print("Contribuição proporcional:", controlador_proporcional(4, 2), "%")
print("Outra condição:", controlador_proporcional(-3, 1.5), "%")
```

**Por que essa versão é melhor?** O resultado não depende de um `Kc` escondido em outra célula. Isso facilita testes, reutilização e comparação de condições operacionais. O sinal da contribuição proporcional depende da convenção adotada para o erro e da ação do controlador.

<div id="secao-8" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  8. Aplicações em Instrumentação e Controle de Processos
</div>

**Instrumentação — transmissor linear de 4–20 mA.** Considere um transmissor de pressão calibrado de 0 a 500 kPa. Admitindo resposta linear, a corrente para uma pressão $P$ é dada por:

$$I=4+16\frac{P}{500}$$

Para $P=250$ kPa, o cálculo manual fornece $I=4+16(250/500)=12$ mA. O intervalo é nominal: valores fora da faixa precisam de interpretação específica conforme o instrumento.

```python
def corrente_transmissor(P, P_min, P_max):
    if P_max <= P_min:
        raise ValueError("A faixa de medição deve ser crescente.")
    return 4 + 16 * (P - P_min) / (P_max - P_min)

print("Corrente:", corrente_transmissor(250, 0, 500), "mA")
```

**Controle de Processos — comparação entre condições.** Em vez de alterar uma variável global, podemos testar diferentes ganhos proporcionais para o mesmo erro.

```python
erro = 4  # °C
for ganho in [1.0, 2.0, 3.0]:
    u = controlador_proporcional(erro, ganho)
    print(f"Kc = {ganho:.1f} %/°C -> uP = {u:.1f} %")
```

<div id="secao-9" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">
  9. Aplicações em Mecânica dos Fluidos e Transferência de Calor
</div>

**Mecânica dos Fluidos — velocidade média em uma tubulação.** A área é $A=\pi D^2/4$ e a velocidade média é $v=Q/A$. Para $Q=0{,}002$ m³/s e $D=0{,}05$ m, obtemos aproximadamente $v=1{,}019$ m/s. Passaremos vazão e diâmetro diretamente à função.

```python
import math

def velocidade_media(Q, D):
    if D <= 0:
        raise ValueError("O diâmetro deve ser positivo.")
    area = math.pi * D**2 / 4
    return Q / area

print(f"Velocidade média: {velocidade_media(0.002, 0.05):.3f} m/s")
```

**Transferência de Calor — condução em parede plana.** Para uma parede homogênea, em regime estacionário e condução unidimensional, $\dot Q=kA(T_1-T_2)/L$. Com $k=0{,}8$ W/(m·K), $A=2$ m², $L=0{,}1$ m e diferença de temperatura de 20 °C, a taxa calculada é 320 W. O sinal informa o sentido adotado para o fluxo.

```python
def taxa_conducao(k, A, L, T1, T2):
    if k <= 0 or A <= 0 or L <= 0:
        raise ValueError("k, A e L devem ser positivos.")
    return k * A * (T1 - T2) / L

print("Taxa de calor:", taxa_conducao(0.8, 2, 0.1, 60, 40), "W")
```

  10. Boas práticas na organização de modelos científicos
</div>

Em programas científicos, boas práticas ajudam a garantir que os resultados possam ser reproduzidos. Prefira funções com entradas explícitas, nomes associados às grandezas físicas, unidades documentadas e saídas obtidas por `return`. Evite alterar variáveis globais sem necessidade. Uma função não deve depender da ordem em que células anteriores foram executadas, sempre que isso puder ser evitado.

```python
def erro_temperatura(T_SP, T_PV):
    """Calcula T_SP - T_PV, em graus Celsius."""
    return T_SP - T_PV

def contribuicao_proporcional(T_SP, T_PV, Kc):
    """Retorna a contribuição proporcional em pontos percentuais."""
    e = erro_temperatura(T_SP, T_PV)
    return Kc * e

assert contribuicao_proporcional(80, 76, 2) == 8
print("Teste aprovado: contribuição proporcional = 8 %")
```
