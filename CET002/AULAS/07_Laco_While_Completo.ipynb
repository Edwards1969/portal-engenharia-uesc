<div id="secao-1" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:22px;"><strong>1. O que é o laço while?</strong></div>

A palavra *while* significa **enquanto**. O programa avalia uma condição lógica antes de cada repetição. Se ela for verdadeira (`True`), executa o bloco indentado; quando se torna falsa (`False`), continua na linha seguinte ao laço.

```python
while condicao:
    instrucao_1
    instrucao_2
```

A condição pode envolver comparações (`<`, `<=`, `>`, `>=`, `==`, `!=`) e operadores lógicos (`and`, `or`, `not`). A **indentação** indica quais instruções pertencem ao laço. Diferentemente de uma sequência de cálculos com número de etapas conhecido, aqui o número de repetições pode depender dos resultados obtidos durante a execução.

<div id="secao-2" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:22px;"><strong>2. A primeira repetição: contador</strong></div>

Considere a necessidade de registrar cinco leituras consecutivas de um instrumento. Antes de associar o código a um equipamento, vamos compreender o mecanismo de repetição.

**Raciocínio manual:** começamos com `leitura = 1`. Enquanto `leitura <= 5`, mostramos o número e acrescentamos uma unidade. Após mostrar a quinta leitura, a variável passa a valer 6 e o laço termina.

````python
leitura = 1
while leitura <= 5:
    print(f"Leitura número {leitura}")
    leitura += 1
print("Registro concluído.")
````

**Interpretação.** A instrução `leitura += 1` equivale a `leitura = leitura + 1`. Ela modifica a variável utilizada na condição. Sem essa atualização, o programa poderia repetir indefinidamente. Observe que a sexta leitura não é exibida: o teste acontece **antes** da execução do bloco.

<div id="secao-3" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:22px;"><strong>3. Condição de parada e segurança do algoritmo</strong></div>

Um laço bem construído precisa de uma **condição de continuidade**, de uma variável que evolua e, em cálculos numéricos, frequentemente de um **limite máximo de iterações**. Essa precaução impede que um erro de formulação mantenha o programa em execução sem fim.

Exemplo: simular a aproximação de uma variável ao valor desejado. A cada etapa, reduzimos pela metade a distância restante. O cálculo termina quando a distância for menor que 0,1 ou quando forem realizadas 20 etapas.

````python
distancia = 10.0
etapa = 0
max_etapas = 20
while distancia > 0.1 and etapa < max_etapas:
    distancia *= 0.5
    etapa += 1
    print(f"Etapa {etapa:2d}: distância = {distancia:.4f}")
if distancia <= 0.1:
    print("Critério de parada atendido.")
else:
    print("Limite de etapas atingido sem convergência.")
````

O operador `and` exige que **as duas condições** sejam verdadeiras para o laço continuar. A estrutura `if` após o `while` informa **por que** a repetição terminou. Isso é especialmente importante em simulações e rotinas de otimização.

<div id="secao-4" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:22px;"><strong>4. while e for: quando usar cada um?</strong></div>

Os dois comandos realizam repetições, mas enfatizam problemas diferentes. Usamos `for` quando queremos percorrer uma sequência conhecida, como 10 amostras ou 24 horários. Usamos `while` quando queremos continuar **até atingir uma condição**, como uma temperatura-alvo ou uma tolerância numérica.

Os exemplos abaixo produzem os mesmos três valores; a diferença está na forma de expressar o algoritmo.

````python
print("Com for:")
for k in range(1, 4):
    print(k)
print("Com while:")
k = 1
while k <= 3:
    print(k)
    k += 1
````

<div id="secao-5" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:22px;"><strong>5. Mecânica dos Fluidos: enchimento de um tanque</strong></div>

**Problema de Engenharia.** Um tanque inicialmente contém 50 L de líquido. A vazão de entrada é 5 L/min e a de saída é 4 L/min. Desejamos determinar o primeiro instante, em passos de 1 minuto, no qual o volume atinge 60 L.

**Modelo matemático.** Admitindo líquido incompressível e vazões constantes, o balanço de volume é

$$
\frac{dV}{dt}=Q_{in}-Q_{out}.
$$

Para um passo de tempo $\Delta t$, a atualização numérica é

$$
V_{k+1}=V_k+(Q_{in}-Q_{out})\Delta t.
$$

**Cálculo manual:** a vazão líquida é $5-4=1$ L/min. Após 1 minuto, $V=51$ L; após 2 minutos, $V=52$ L. Como são necessários 10 L adicionais, o volume-alvo é atingido em 10 minutos. Vamos reproduzir esse raciocínio com `while`.

````python
volume = 50.0       # L
q_entrada = 5.0      # L/min
q_saida = 4.0        # L/min
volume_alvo = 60.0   # L
dt = 1.0             # min
tempo = 0.0
max_passos = 100
passo = 0

if q_entrada <= q_saida and volume < volume_alvo:
    print("O volume não aumentará: verifique as vazões.")
else:
    while volume < volume_alvo and passo < max_passos:
        volume += (q_entrada - q_saida) * dt
        tempo += dt
        passo += 1
    print(f"Volume final: {volume:.1f} L")
    print(f"Tempo simulado: {tempo:.1f} min")
    print(f"Alvo atingido? {volume >= volume_alvo}")
````

**Interpretação.** O resultado é 60 L em 10 min, de acordo com o cálculo manual. Se o passo de tempo fosse maior, o laço poderia ultrapassar o volume-alvo antes de parar. Em aplicações reais, a dinâmica de válvulas, sensores, limites do tanque e procedimentos de segurança exigem modelos e sistemas de proteção adicionais; este exemplo representa apenas o balanço de volume.

<div id="secao-6" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:22px;"><strong>6. Instrumentação e Controle: acompanhamento de temperatura</strong></div>

**Problema de Engenharia.** Durante o aquecimento de um processo, um sensor registra a temperatura a cada minuto. Para introduzir a lógica do monitoramento, suponha que a temperatura aumente 2 °C por minuto, partindo de 30 °C. Queremos saber quando a leitura alcançará 40 °C.

**Modelo didático:**

$$
T_{k+1}=T_k+r\Delta t,
$$

com $r=2$ °C/min e $\Delta t=1$ min. **Cálculo manual:** as leituras são 30, 32, 34, 36, 38 e 40 °C; portanto, são necessários 5 minutos. O código simula essas leituras; **não controla um equipamento real**.

````python
temperatura = 30.0   # °C
temperatura_alvo = 40.0
rampa = 2.0          # °C/min
dt = 1.0             # min
tempo = 0.0
max_passos = 30
passo = 0

print(f"t = {tempo:4.1f} min | T = {temperatura:4.1f} °C")
while temperatura < temperatura_alvo and passo < max_passos:
    temperatura += rampa * dt
    tempo += dt
    passo += 1
    print(f"t = {tempo:4.1f} min | T = {temperatura:4.1f} °C")
print(f"Alvo alcançado: {temperatura >= temperatura_alvo}")
````

**Interpretação.** O programa termina quando a temperatura simulada chega a 40 °C. Em Instrumentação, é importante distinguir o **valor medido** do **valor calculado por um modelo**. Em Controle de Processos Aplicado, o monitoramento real requer aquisição de sinais, tratamento de falhas de sensor e proteções independentes do código didático.

<div id="secao-7" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:22px;"><strong>7. Transferência de Calor: resfriamento de um corpo</strong></div>

**Problema de Engenharia.** Uma peça inicialmente a 80 °C resfria em um ambiente a 25 °C. Desejamos estimar, em passos de 1 minuto, quando sua temperatura será inferior ou igual a 35 °C.

**Modelo matemático.** Pela lei de resfriamento de Newton, com coeficiente global simplificado $k$,

$$
\frac{dT}{dt}=-k(T-T_{amb}).
$$

Usando Euler explícito:

$$
T_{k+1}=T_k-k(T_k-T_{amb})\Delta t.
$$

Para $k=0{,}2$ por minuto e $\Delta t=1$ min, a primeira etapa é $T_1=80-0{,}2(80-25)=69$ °C. A segunda é $T_2=69-0{,}2(69-25)=60{,}2$ °C. O `while` continuará até alcançar o limite.

````python
temperatura = 80.0  # °C
t_ambiente = 25.0    # °C
t_limite = 35.0      # °C
k = 0.2              # 1/min
dt = 1.0             # min
tempo = 0.0
passo = 0
max_passos = 100

if not 0 < k * dt <= 1:
    print("Revise k e dt para este exemplo numérico.")
else:
    while temperatura > t_limite and passo < max_passos:
        temperatura -= k * (temperatura - t_ambiente) * dt
        tempo += dt
        passo += 1
        print(f"t = {tempo:4.1f} min | T = {temperatura:6.2f} °C")
    print(f"Limite alcançado: {temperatura <= t_limite}")
````

**Interpretação.** O corpo se aproxima da temperatura ambiente progressivamente. A taxa de resfriamento diminui porque a diferença entre a temperatura do corpo e a do ambiente fica menor. A equação representa um **modelo concentrado**; sua adequação depende das propriedades da peça e das condições de troca térmica.

<div id="secao-8" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:22px;"><strong>8. Modelagem e Simulação: método iterativo</strong></div>

**Problema de Engenharia.** Em balanços não lineares, é comum que a incógnita apareça dos dois lados de uma equação. Considere, apenas para estudar o algoritmo,

$$
x=\sqrt{10-x}.
$$

Podemos obter uma aproximação sucessiva por

$$
x_{k+1}=\sqrt{10-x_k}.
$$

**Cálculo manual:** partindo de $x_0=2$, temos $x_1=\sqrt{8}\approx 2{,}8284$ e $x_2=\sqrt{10-2{,}8284}\approx 2{,}6780$. Continuaremos até que a diferença absoluta entre duas aproximações seja inferior a $10^{-6}$. Usaremos também um máximo de 100 iterações.

````python
import math

x = 2.0
tolerancia = 1e-6
max_iteracoes = 100
iteracao = 0
erro = float("inf")

while erro > tolerancia and iteracao < max_iteracoes:
    if 10 - x < 0:
        print("Iteração interrompida: raiz quadrada sem valor real.")
        break
    x_novo = math.sqrt(10 - x)
    erro = abs(x_novo - x)
    x = x_novo
    iteracao += 1

print(f"Aproximação: {x:.6f}")
print(f"Erro entre iterações: {erro:.2e}")
print(f"Iterações: {iteracao}")
print(f"Convergiu pelo critério adotado? {erro <= tolerancia}")
````

**Interpretação.** A equação pode ser verificada elevando ambos os lados ao quadrado: $x^2+x-10=0$, cuja raiz positiva é $(-1+\sqrt{41})/2\approx 2{,}70156$. A diferença entre iterações é um **critério de parada**, mas nem sempre garante, sozinha, que um método convergiu para a solução correta. Em problemas industriais, também verificamos o resíduo da equação e a plausibilidade física da resposta.

<div id="secao-9" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:22px;"><strong>9. Controle de Qualidade e Processos de Fabricação</strong></div>

**Problema de Engenharia.** Em uma linha de fabricação, amostras são avaliadas sequencialmente. Para um exercício de Controle de Qualidade, consideraremos uma lista de medições de concentração e interromperemos a inspeção na primeira leitura fora da faixa didática de 9,5 a 10,5 g/L.

**Raciocínio manual:** as três primeiras leituras (10,1; 10,3; 9,8 g/L) atendem ao critério. A quarta (10,7 g/L) está fora da faixa, de modo que o procedimento de verificação simulado deve parar nessa amostra. Este é um exemplo de **triagem por limites**, não de controle estatístico de processo.

````python
amostras = [10.1, 10.3, 9.8, 10.7, 10.0]
limite_inferior = 9.5
limite_superior = 10.5
indice = 0

while indice < len(amostras):
    valor = amostras[indice]
    print(f"Amostra {indice + 1}: {valor:.1f} g/L")
    if not (limite_inferior <= valor <= limite_superior):
        print("Leitura fora da faixa didática. Verificação interrompida.")
        break
    indice += 1
else:
    print("Todas as amostras avaliadas estão na faixa.")
````

**Interpretação.** O comando `break` encerra imediatamente o laço. O `else` associado ao `while` é executado apenas quando o laço termina **sem `break`**. Em uma planta real, a investigação de uma não conformidade envolve procedimentos documentados, incerteza de medição, rastreabilidade e critérios de aceitação apropriados ao produto.

<div id="secao-10" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:22px;"><strong>10. Ciência dos Materiais: tratamento térmico simplificado</strong></div>

**Problema de Engenharia.** Durante uma etapa idealizada de aquecimento de uma amostra, desejamos registrar os instantes em que a temperatura simulada ultrapassa 100 °C. Admitiremos uma rampa constante de 15 °C/min, começando em 25 °C, até atingir pelo menos 130 °C.

**Modelo:** $T_{k+1}=T_k+15\Delta t$, com $\Delta t=1$ min. Manualmente, a sequência é 25, 40, 55, 70, 85, 100, 115 e 130 °C. O primeiro valor **acima** de 100 °C ocorre no sexto minuto. Não estamos modelando transformações de fase, cinética ou propriedades de um material específico.

````python
temperatura = 25.0  # °C
temperatura_final = 130.0
rampa = 15.0        # °C/min
tempo = 0
max_passos = 20

while temperatura < temperatura_final and tempo < max_passos:
    temperatura += rampa
    tempo += 1
    if temperatura > 100:
        print(f"t = {tempo} min: T = {temperatura:.1f} °C (acima de 100 °C)")
print(f"Fim da simulação: {temperatura:.1f} °C em {tempo} min")
````

**Interpretação.** A estrutura `if` dentro do `while` permite tomar uma decisão **em cada etapa** da repetição. Em Ciência dos Materiais, a mesma organização computacional pode ser aplicada, com modelos adequados, ao acompanhamento de históricos térmicos, processos de secagem, sinterização e tratamentos térmicos.

<div id="secao-11" style="background-color:#f1faff; border-left:6px solid #0074B7; color:#003366; border-radius:8px; padding:15px 20px; font-size:22px;"><strong>11. Cuidados frequentes com o while</strong></div>

Ao escrever um laço, verifique três perguntas: **(1)** a condição inicial está correta? **(2)** alguma variável muda a cada repetição? **(3)** existe uma maneira segura de interromper a execução caso o resultado esperado não seja alcançado?

Também é importante lembrar que `while` testa a condição **antes** do bloco; que `break` interrompe o laço imediatamente; e que `continue` inicia a próxima verificação da condição, podendo causar repetições infinitas se for usado antes de atualizar o contador.

O exemplo seguinte mostra `continue` em uma lista de leituras: valores negativos, usados aqui como indicação didática de leitura inválida, são ignorados, mas o índice é atualizado **antes** da decisão.

````python
leituras = [12.0, -1.0, 11.8, 12.2]
i = 0
while i < len(leituras):
    valor = leituras[i]
    i += 1  # Atualização antes de um possível continue
    if valor < 0:
        print("Leitura inválida ignorada.")
        continue
    print(f"Leitura válida: {valor:.1f}")
````
