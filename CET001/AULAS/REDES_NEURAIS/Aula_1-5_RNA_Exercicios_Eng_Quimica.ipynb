<div id="secao-1" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">1. Problema de Engenharia Química</div>

Um Engenheiro Químico deseja prever a __eficiência de conversão de um reator catalítico__ ( $\eta$, variando de 0 a 1) a partir de três variáveis de processo:

- $T$: Temperatura de operação (em $^oC$)
- $C_A$: Concentração inicial do reagente A (mol/L)
- $P$: Pressão de operação (bar)

Foi proposta uma rede neural __feedforward__ com:

- 3 entradas ($T$, $C_A$, $P$)
- 3 camadas ocultas (cada uma com 3 neurônios)
- Função de ativação sigmoide
- 1 neurônio de saída (eficiência prevista $\eta$)

Sabendo-se que os pesos e termos de viés (*bias*) são fornecidos no exercício, __implemente o cálculo de saída da rede usando operações matriciais__ (método algébrico) em Python com NumPy.

**Observação:** os pesos fornecidos são ilustrativos. Não há dados experimentais nem procedimento de calibração documentados neste exercício; portanto, o resultado não deve ser interpretado como uma previsão validada de um reator real.

<div id="secao-2" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">2. Notação e organização matricial</div>

__Rotulando o problema__:

Vamos identificar as variáveis e os parâmetros usando a seguinte sequência:

- __Entradas rotuladas__ com $x_1$, $x_2$, $x_3$;
- __Pesos identificados__ com $\omega_{ij}^{(1)}$ para indicar a camada;
- __Bias__ indicados como $b_j^{(1)}$, $b_j^{(2)}$, etc;

Essas são notações para conexões entre camadas mostrando os valores reais dos pesos do exercício.

Sendo assim, teremos:

- Entrada: $(x_1,x_2,x_3)=(350;\,1{,}2;\,5{,}0)$, correspondendo a $(T,C_A,P)$
- __Primeira camada oculta__ (3 neurônios):
  
  Pesos $\omega_{ij}^{(1)}$ organizados na matriz $W_1$ e bias $b_j^{(1)}$;
  
- __Segunda camada oculta__ (3 neurônios):
   
    Pesos $\omega_{ij}^{(2)}$ organizados na matriz $W_2$ e bias $b_j^{(2)}$;

- __Terceira camada oculta__ (3 neurônios):
   
    Pesos $\omega_{ij}^{(3)}$ organizados na matriz $W_3$ e bias $b_j^{(3)}$;

- __Saída__ (1 neurônio):
Peso $W_4$, bias $b^{(4)}$

Adotamos a convenção de que cada **linha** de $W_\ell$ representa as conexões de um neurônio da camada anterior e cada **coluna** representa um neurônio da camada seguinte. Assim, $W_1,W_2,W_3\in\mathbb{R}^{3\times3}$ e $W_4\in\mathbb{R}^{3\times1}$. Os vetores de viés das camadas ocultas têm dimensão $1\times3$.

<div id="secao-3" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">3. Arquitetura e equações da rede neural</div>

A arquitetura da rede é representada esquematicamente na figura a seguir:

Para uma rede *feedforward*, a propagação da informação é feita da entrada até a saída, sem atualização dos pesos. Com a função sigmoide

$$\sigma(z)=\frac{1}{1+e^{-z}},$$

as operações são

$$\begin{aligned}
\mathbf{a}^{(0)}&=\mathbf{x},\\
\mathbf{z}^{(1)}&=\mathbf{x}W_1+B_1,&\mathbf{a}^{(1)}&=\sigma(\mathbf{z}^{(1)}),\\
\mathbf{z}^{(2)}&=\mathbf{a}^{(1)}W_2+B_2,&\mathbf{a}^{(2)}&=\sigma(\mathbf{z}^{(2)}),\\
\mathbf{z}^{(3)}&=\mathbf{a}^{(2)}W_3+B_3,&\mathbf{a}^{(3)}&=\sigma(\mathbf{z}^{(3)}),\\
\mathbf{z}^{(4)}&=\mathbf{a}^{(3)}W_4+B_4,&\widehat{\eta}&=\sigma(\mathbf{z}^{(4)}).
\end{aligned}$$

A sigmoide é aplicada elemento a elemento. Como o resultado está entre zero e um, a eficiência calculada pode ser apresentada em forma adimensional ou percentual. **Este exercício realiza somente a propagação direta (*forward pass*), não o treinamento da rede.**

<div id="secao-4" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">4. Implementação em Python com NumPy</div>

A implementação utiliza as matrizes de pesos e os vetores de viés fornecidos no enunciado.

```python
import numpy as np

# Função sigmoide
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# =========================
# Entrada
# =========================
x1, x2, x3 = 350, 1.2, 5.0
entrada = np.array([[x1, x2, x3]])

# =========================
# Pesos e Bias - Camada Oculta 1
# =========================
w11, w12, w13 = 0.12, -0.25, 0.33
w21, w22, w23 = -0.45, 0.67, -0.12
w31, w32, w33 = 0.56, 0.11, -0.78

b1_1, b1_2, b1_3 = 0.10, -0.05, 0.20

W1 = np.array([[w11, w12, w13],
               [w21, w22, w23],
               [w31, w32, w33]])
B1 = np.array([[b1_1, b1_2, b1_3]])

# =========================
# Pesos e Bias - Camada Oculta 2
# =========================
w11_2, w12_2, w13_2 = -0.32, 0.25, 0.40
w21_2, w22_2, w23_2 = 0.28, -0.14, 0.33
w31_2, w32_2, w33_2 = -0.56, 0.22, 0.15

b2_1, b2_2, b2_3 = -0.10, 0.05, -0.15

W2 = np.array([[w11_2, w12_2, w13_2],
               [w21_2, w22_2, w23_2],
               [w31_2, w32_2, w33_2]])
B2 = np.array([[b2_1, b2_2, b2_3]])

# =========================
# Pesos e Bias - Camada Oculta 3
# =========================
w11_3, w12_3, w13_3 = 0.45, -0.20, 0.30
w21_3, w22_3, w23_3 = -0.25, 0.60, -0.50
w31_3, w32_3, w33_3 = 0.15, -0.40, 0.55

b3_1, b3_2, b3_3 = 0.05, -0.05, 0.10

W3 = np.array([[w11_3, w12_3, w13_3],
               [w21_3, w22_3, w23_3],
               [w31_3, w32_3, w33_3]])
B3 = np.array([[b3_1, b3_2, b3_3]])

# =========================
# Pesos e Bias - Camada de Saída
# =========================
w11_4, w21_4, w31_4 = 0.40, -0.30, 0.25

b4_1 = 0.05

W4 = np.array([[w11_4],
               [w21_4],
               [w31_4]])
B4 = np.array([[b4_1]])

# =========================
# Forward Pass
# =========================
# Camada oculta 1
z1 = np.dot(entrada, W1) + B1
a1 = sigmoid(z1)

# Camada oculta 2
z2 = np.dot(a1, W2) + B2
a2 = sigmoid(z2)

# Camada oculta 3
z3 = np.dot(a2, W3) + B3
a3 = sigmoid(z3)

# Saída
z4 = np.dot(a3, W4) + B4
saida = sigmoid(z4)

# =========================
# Resultado
# =========================
print(f"Saída calculada da rede (adimensional): {saida[0,0]:.4f}")
print(f"Saída calculada da rede (percentual): {100*saida[0,0]:.2f}%")
```

<div id="secao-5" style="background-color:#f1faff; border-left:6px solid #0074B7; text-align:left; color:#003366; font-size:22px; font-weight:bold; border-radius:8px; padding:15px 20px; margin-top:20px;">5. Interpretação e limitações</div>

O código calcula a saída da arquitetura e dos pesos especificados. Isso demonstra como a multiplicação de matrizes permite encadear as camadas da rede neural. **Não demonstra que a eficiência real do reator seja igual ao valor calculado.**

**Atenção às escalas:** a temperatura foi introduzida diretamente como $350$, enquanto concentração e pressão são numericamente muito menores. Com os pesos fornecidos, isso gera valores de grande magnitude na primeira camada, fazendo algumas sigmoides se aproximarem de $0$ ou $1$ (saturação). Em uma aplicação real, seria necessário definir e documentar um procedimento de escalonamento das entradas, além de treinar e validar os parâmetros com dados experimentais. Não alteramos arbitrariamente as entradas ou os pesos, pois isso mudaria o exercício proposto.

Como extensão, o estudante pode calcular manualmente $\mathbf{z}^{(1)}$ e $\mathbf{a}^{(1)}$ e comparar os resultados com as matrizes obtidas pelo NumPy.
