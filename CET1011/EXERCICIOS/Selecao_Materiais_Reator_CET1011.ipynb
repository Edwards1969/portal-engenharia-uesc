### 1. Dados geométricos e operacionais do reator

A geometria do casco cilíndrico e a diferença de pressão entre o interior e o exterior do equipamento constituem os parâmetros fundamentais para o dimensionamento mecânico preliminar.

Para o problema em estudo, serão consideradas as seguintes condições:

| Grandeza | Símbolo | Valor | Unidade |
|:---|:---:|---:|:---:|
| Diâmetro interno do casco | $D_i$ | 1200 | mm |
| Comprimento da região cilíndrica | $L_c$ | 2400 | mm |
| Diferença de pressão de projeto | $P$ | 1,50 | MPa |

O diâmetro interno representa a dimensão transversal da região cilíndrica em contato com o fluido de processo, enquanto o comprimento corresponde à extensão axial da região do casco considerada no modelo matemático.

A diferença de pressão de projeto representa a diferença entre a pressão interna e a pressão externa do equipamento:

$$
P=P_{\mathrm{int}}-P_{\mathrm{ext}}
$$

Essa diferença de pressão produz esforços mecânicos na parede do casco, originando tensões que deverão ser suportadas pelo material de construção.

Para compreender a ordem de grandeza das condições operacionais, podemos converter a pressão de projeto para outras unidades:

$$
P=1,50\ \mathrm{MPa}
$$

$$
P=1,50\times10^6\ \mathrm{Pa}
$$

$$
P=15,0\ \mathrm{bar}
$$

Portanto, o casco cilíndrico deverá ser dimensionado preliminarmente para suportar uma diferença de pressão de projeto de 15 bar.

É importante observar que o diâmetro interno e a pressão de projeto influenciam diretamente a espessura mínima necessária, enquanto o comprimento da região cilíndrica influencia a quantidade de material necessária e, consequentemente, a massa e o custo de matéria-prima do casco.

### 2. Propriedades dos materiais candidatos

A seleção de materiais para equipamentos industriais constitui um problema de engenharia que envolve diferentes critérios de avaliação.

Um material pode apresentar elevada resistência mecânica, permitindo a utilização de paredes menos espessas, mas possuir um custo de aquisição relativamente elevado. Outro material pode apresentar um custo significativamente menor, mas não possuir características adequadas às condições químicas do processo.

Para este estudo, serão considerados três materiais hipotéticos, denominados Material A, Material B e Material C, cujas propriedades são apresentadas na tabela a seguir.

| Propriedade | Material A | Material B | Material C |
|:---|---:|---:|---:|
| Tensão admissível (MPa) | 120 | 150 | 180 |
| Massa específica (kg/m³) | 7850 | 8000 | 7900 |
| Custo unitário (R$/kg) | 8,00 | 24,00 | 35,00 |
| Compatibilidade química | Incompatível | Compatível | Compatível |

Os valores apresentados são hipotéticos e foram estabelecidos exclusivamente para o desenvolvimento deste estudo didático.

#### 2.1. Tensão admissível

A tensão admissível, representada por $S_{\mathrm{adm}}$, corresponde ao limite de tensão estabelecido para a avaliação mecânica do material nas condições consideradas pelo modelo.

O critério mecânico simplificado será representado por:

$$
\sigma_{\theta}\leq S_{\mathrm{adm}}
$$

em que $\sigma_{\theta}$ representa a tensão circunferencial desenvolvida na parede cilíndrica.

Para uma mesma pressão de projeto e um mesmo diâmetro interno, materiais com maiores tensões admissíveis permitem, segundo o modelo utilizado, a adoção de menores espessuras de parede.

Entretanto, uma maior tensão admissível não significa, necessariamente, que o material seja a alternativa mais adequada ao projeto.

#### 2.2. Massa específica

A massa específica, representada pela letra grega $\rho$, estabelece a relação entre a massa e o volume de um material:

$$
\rho=\frac{m}{V}
$$

Consequentemente, a massa de material necessária à construção da parede cilíndrica poderá ser determinada por:

$$
m=\rho V_{\mathrm{mat}}
$$

em que $V_{\mathrm{mat}}$ representa o volume de matéria-prima necessário à construção da parede.

A massa do casco dependerá, portanto, tanto da massa específica do material quanto da espessura necessária para atender às condições mecânicas do problema.

Assim, um material com maior massa específica poderá resultar em um casco mais leve, caso permita a utilização de uma parede suficientemente menos espessa.

#### 2.3. Custo unitário da matéria-prima

O custo unitário representa o valor de aquisição de um quilograma de determinado material.

Conhecendo a massa necessária à construção da parede cilíndrica, o custo estimado de matéria-prima poderá ser determinado por:

$$
C=m\,c_{\mathrm{kg}}
$$

em que:

- $C$ representa o custo total estimado de matéria-prima, em reais;
- $m$ representa a massa da parede cilíndrica, em quilogramas;
- $c_{\mathrm{kg}}$ representa o custo unitário do material, em R$/kg.

A análise econômica deverá considerar simultaneamente a quantidade de material necessária e seu respectivo custo unitário.

Consequentemente, a alternativa de menor massa não será necessariamente a alternativa de menor custo.

#### 2.4. Compatibilidade química

Além das propriedades mecânicas, físicas e econômicas, a seleção do material deverá considerar sua compatibilidade química com o fluido de processo.

Para as condições hipotéticas estabelecidas neste problema:

- O Material A não atende ao requisito de compatibilidade química;
- O Material B atende ao requisito de compatibilidade química;
- O Material C atende ao requisito de compatibilidade química.

A incompatibilidade química constitui uma restrição técnica eliminatória, independentemente da resistência mecânica ou do custo apresentado pelo material.

A utilização de um material incompatível com o meio de processo pode provocar fenômenos de degradação, corrosão, contaminação do produto e comprometimento da integridade estrutural do equipamento.

Portanto, a comparação econômica destinada à seleção final deverá ser realizada exclusivamente entre os materiais tecnicamente admissíveis.

### 3. Espessuras comerciais disponíveis

Em um problema de dimensionamento, o modelo matemático pode fornecer qualquer valor de espessura dentro do domínio contínuo dos números reais.

Entretanto, em uma situação industrial, as chapas metálicas são normalmente disponibilizadas em determinadas espessuras comerciais.

Para o problema em estudo, será considerado o seguinte conjunto hipotético de espessuras disponíveis:

$$
\mathcal{T}=\{5,\ 6,\ 8,\ 10,\ 12,\ 16\}\ \mathrm{mm}
$$

A espessura mínima calculada pelo modelo matemático será representada por:

$$
t_{\min}
$$

A espessura comercial efetivamente selecionada será representada por:

$$
t_{\mathrm{adot}}
$$

A seleção deverá atender à seguinte condição:

$$
t_{\mathrm{adot}}\geq t_{\min}
$$

Entre todas as espessuras comerciais que atendem a essa condição, deverá ser selecionada a menor delas:

$$
\boxed{
t_{\mathrm{adot}}=
\min\left\{
t\in\mathcal{T}:t\geq t_{\min}
\right\}
}
$$

Por exemplo, suponha que o dimensionamento preliminar de determinado equipamento forneça uma espessura mínima de 9,20 mm.

Considerando o conjunto de espessuras comerciais disponíveis, uma chapa de 8 mm não poderá ser utilizada, pois sua espessura é inferior ao valor mínimo calculado.

A menor espessura comercial que atende à condição estabelecida será:

$$
t_{\mathrm{adot}}=10,00\ \mathrm{mm}
$$

Embora uma chapa de 12 mm também satisfaça o critério mecânico simplificado, sua utilização implicaria um aumento da quantidade de matéria-prima, da massa e do custo do equipamento.

A seleção da espessura comercial representa, portanto, um primeiro problema discreto de otimização, no qual se deseja identificar a menor espessura disponível que satisfaça a restrição mecânica estabelecida.

Essa lógica será posteriormente implementada no programa Python, permitindo que a seleção seja realizada automaticamente para cada material candidato.

### 4. Organização dos dados de entrada em Python

Uma das primeiras etapas do desenvolvimento de uma aplicação computacional consiste em transformar as informações do problema de engenharia em estruturas de dados que possam ser manipuladas pelo programa.

Neste estudo, utilizaremos três recursos fundamentais da linguagem Python:

- Variáveis numéricas para armazenar os parâmetros geométricos e operacionais;
- Uma lista para armazenar as espessuras comerciais disponíveis;
- Um dicionário para armazenar as propriedades dos materiais candidatos.

A utilização de um dicionário permitirá associar cada material às suas respectivas propriedades mecânicas, físicas, econômicas e químicas.

Essa organização evitará a necessidade de escrever individualmente as mesmas equações para cada material, permitindo que os cálculos sejam realizados posteriormente por meio de uma estrutura de repetição.

Inicialmente, importaremos a biblioteca `math`, que será utilizada no cálculo do volume da parede cilíndrica, e definiremos todos os dados de entrada do problema.

````python
import math

# ==================================================
# 1. DADOS GEOMETRICOS E OPERACIONAIS
# ==================================================

D_mm = 1200.0       # Diametro interno (mm)
L_mm = 2400.0       # Comprimento cilindrico (mm)
P_MPa = 1.50        # Diferenca de pressao (MPa)


# ==================================================
# 2. ESPESSURAS COMERCIAIS DISPONIVEIS
# ==================================================

espessuras_mm = [5, 6, 8, 10, 12, 16]


# ==================================================
# 3. PROPRIEDADES DOS MATERIAIS
# ==================================================

materiais = {
    "A": {
        "S_MPa": 120.0,
        "rho": 7850.0,
        "preco": 8.0,
        "compativel": False
    },
    
    "B": {
        "S_MPa": 150.0,
        "rho": 8000.0,
        "preco": 24.0,
        "compativel": True
    },
    
    "C": {
        "S_MPa": 180.0,
        "rho": 7900.0,
        "preco": 35.0,
        "compativel": True
    }
}


# ==================================================
# 4. CONVERSAO DAS UNIDADES GEOMETRICAS
# ==================================================

D_m = D_mm / 1000.0
L_m = L_mm / 1000.0


# ==================================================
# 5. TOLERANCIA NUMERICA
# ==================================================

tol = 1e-10
````

### 5. Compreendendo a estrutura de dados utilizada

Observe que as propriedades dos materiais foram organizadas em um dicionário denominado `materiais`.

Cada material possui uma chave de identificação (`"A"`, `"B"` ou `"C"`), associada a um segundo dicionário contendo suas propriedades.

As chaves utilizadas para identificar essas propriedades são:

| Chave | Significado | Unidade ou tipo |
|:---|:---|:---:|
| `S_MPa` | Tensão admissível | MPa |
| `rho` | Massa específica | kg/m³ |
| `preco` | Custo unitário da matéria-prima | R$/kg |
| `compativel` | Compatibilidade química | Booleano |

A propriedade `compativel` foi representada por uma variável lógica, também denominada variável booleana.

Em Python, uma variável booleana pode assumir dois valores:

- `True`: condição verdadeira;
- `False`: condição falsa.

Essa representação será particularmente importante na etapa de avaliação das restrições técnicas, pois permitirá verificar automaticamente se determinado material atende ao requisito de compatibilidade química.

Para compreender como as informações são acessadas, vamos inicialmente consultar as propriedades do Material B.

````python
# Acessando as propriedades do Material B

print("PROPRIEDADES DO MATERIAL B")
print("-" * 45)

print(f"Tensao admissivel: {materiais['B']['S_MPa']:.2f} MPa")

print(f"Massa especifica: {materiais['B']['rho']:.2f} kg/m3")

print(f"Custo unitario: R$ {materiais['B']['preco']:.2f}/kg")

print(f"Compatibilidade quimica: {materiais['B']['compativel']}")
````

### 6. Acessando automaticamente as propriedades dos materiais

O procedimento anterior permite acessar individualmente as propriedades de determinado material.

Entretanto, em um problema de engenharia que envolve a comparação de diferentes alternativas, não é conveniente escrever repetidamente as mesmas instruções para cada material candidato.

Podemos utilizar uma estrutura de repetição `for`, associada ao método `items()`, para percorrer automaticamente todos os materiais armazenados no dicionário.

A instrução:

```python
for nome, dados in materiais.items():
```

permite acessar simultaneamente:

- `nome`: a chave de identificação do material;
- `dados`: o dicionário contendo as propriedades daquele material.

Na primeira repetição, serão acessadas as propriedades do Material A. Na segunda, as propriedades do Material B. Na terceira, as propriedades do Material C.

Esse procedimento constitui a base do algoritmo que será utilizado nas próximas etapas para realizar automaticamente os cálculos de dimensionamento, massa e custo de todos os materiais candidatos.

Vamos inicialmente utilizar essa estrutura de repetição apenas para apresentar os dados de entrada em uma tabela organizada.

````python
# Apresentacao dos dados geometricos e operacionais

print("DADOS GEOMETRICOS E OPERACIONAIS")
print("=" * 70)

print(f"Diametro interno: {D_mm:.2f} mm")
print(f"Comprimento cilindrico: {L_mm:.2f} mm")
print(f"Diferenca de pressao: {P_MPa:.2f} MPa")

print(f"\nEspessuras comerciais (mm): {espessuras_mm}")


# Apresentacao das propriedades dos materiais

print("\nPROPRIEDADES DOS MATERIAIS")
print("=" * 70)

print(
    f"{'Material':>8} | "
    f"{'S (MPa)':>9} | "
    f"{'rho (kg/m3)':>12} | "
    f"{'Preco (R$/kg)':>13} | "
    f"{'Compatibilidade':>15}"
)

print("-" * 70)

for nome, dados in materiais.items():

    quimica = (
        "Compativel" if dados["compativel"]
        else "Incompativel"
    )

    print(
        f"{nome:>8} | "
        f"{dados['S_MPa']:9.2f} | "
        f"{dados['rho']:12.2f} | "
        f"{dados['preco']:13.2f} | "
        f"{quimica:>15}"
    )

print("=" * 70)
````

A resolução de um problema de engenharia por meio da computação científica exige, inicialmente, a construção de um modelo matemático capaz de representar os fenômenos físicos relevantes para o sistema em estudo.

No caso de um reator químico submetido à pressão interna, a diferença de pressão entre o interior e o exterior do equipamento produz esforços mecânicos sobre suas paredes. Esses esforços originam tensões que deverão ser suportadas pelo material de construção.

O dimensionamento mecânico consiste, portanto, em determinar uma geometria capaz de suportar as condições estabelecidas para o equipamento, respeitando os limites admissíveis do material.

Entretanto, o aumento da espessura da parede, embora contribua para a redução das tensões mecânicas, também implica um aumento da quantidade de matéria-prima necessária, da massa do equipamento e dos custos associados à sua construção.

Existe, portanto, uma relação entre o dimensionamento mecânico, as propriedades dos materiais e os aspectos econômicos do projeto.

Para compreender essa relação, desenvolveremos inicialmente as equações que fundamentam nosso modelo matemático. Posteriormente, essas equações serão implementadas em Python, permitindo avaliar diferentes alternativas de materiais de maneira sistemática.

O modelo matemático será construído a partir de três fundamentos:

- Equilíbrio mecânico da parede cilíndrica submetida à pressão interna;
- Relações geométricas utilizadas na determinação do volume de material;
- Relações físicas e econômicas utilizadas na determinação da massa e do custo de matéria-prima.

É importante compreender que o programa computacional não substitui esses fundamentos. Ele constitui uma ferramenta para implementar, resolver e investigar o modelo matemático desenvolvido.

### 1. Hipóteses adotadas no desenvolvimento do modelo

A representação matemática de um equipamento industrial exige a definição de hipóteses que permitam selecionar os fenômenos físicos relevantes e estabelecer os limites de aplicação do modelo.

Um reator químico real apresenta uma geometria relativamente complexa, constituída por casco, tampos, bocais, flanges, juntas soldadas, suportes e componentes internos.

Além disso, durante sua operação, o equipamento pode estar submetido simultaneamente a esforços mecânicos, térmicos, químicos e estruturais.

A consideração de todos esses fenômenos exigiria modelos matemáticos e procedimentos de dimensionamento significativamente mais complexos.

Como nosso objetivo consiste em realizar uma avaliação preliminar de materiais, serão adotadas as seguintes hipóteses:

1. O casco possui geometria cilíndrica circular;
2. A espessura da parede é uniforme ao longo da região analisada;
3. O material é considerado homogêneo e isotrópico;
4. A diferença de pressão é considerada uniforme e estática;
5. A parede cilíndrica será inicialmente tratada como uma parede delgada;
6. O dimensionamento preliminar será fundamentado na tensão circunferencial;
7. Os efeitos de concentrações localizadas de tensão serão desconsiderados;
8. Não serão consideradas sobre-espessuras de corrosão, tolerâncias de fabricação ou fatores de eficiência das juntas soldadas;
9. O cálculo da massa será restrito à parede cilíndrica do casco;
10. A análise econômica considerará exclusivamente o custo de matéria-prima.

Essas hipóteses permitem reduzir a complexidade do problema e estabelecer um modelo matemático adequado aos objetivos deste estudo.

Entretanto, é fundamental compreender que as simplificações adotadas também estabelecem os limites de validade dos resultados obtidos.

A solução desenvolvida neste material deverá ser interpretada como uma avaliação preliminar de engenharia, não constituindo um procedimento completo de dimensionamento ou uma autorização para fabricação de um vaso de pressão industrial.

### 2. Hipótese de vaso de pressão de parede delgada

A distribuição das tensões desenvolvidas na parede de um vaso de pressão depende, entre outros fatores, da relação entre sua espessura e suas dimensões geométricas.

Quando a espessura da parede é suficientemente pequena em relação ao raio interno do equipamento, torna-se possível utilizar uma aproximação denominada teoria de vasos de pressão de parede delgada.

Nessa aproximação, as tensões de membrana são consideradas aproximadamente uniformes ao longo da espessura da parede, e a tensão radial pode ser desprezada em relação às principais tensões de membrana.

Um critério geométrico usual para avaliar a aplicabilidade dessa aproximação é:

$$
\frac{t}{r_i}\leq 0,10
$$

em que:

- $t$ representa a espessura da parede;
- $r_i$ representa o raio interno do casco cilíndrico.

Considerando que:

$$
r_i=\frac{D_i}{2}
$$

o critério também pode ser escrito como:

$$
\boxed{
\frac{t}{D_i}\leq 0,05
}
$$

ou, de maneira equivalente:

$$
\boxed{
\frac{D_i}{t}\geq 20
}
$$

Essas três expressões representam o mesmo critério geométrico.

Para o equipamento em estudo, o diâmetro interno é:

$$
D_i=1200\ \mathrm{mm}
$$

Portanto, seu raio interno corresponde a:

$$
r_i=\frac{1200}{2}=600\ \mathrm{mm}
$$

Considerando a maior espessura presente no conjunto de espessuras comerciais disponíveis:

$$
t=16\ \mathrm{mm}
$$

obtemos:

$$
\frac{t}{r_i}=\frac{16}{600}
$$

$$
\frac{t}{r_i}=0,02667
$$

Como:

$$
0,02667<0,10
$$

verificamos que mesmo a maior espessura comercial disponível satisfaz o critério geométrico usual de parede delgada.

Consequentemente, todas as espessuras comerciais consideradas neste problema encontram-se dentro desse limite geométrico.

Essa verificação fornece uma justificativa inicial para a utilização do modelo simplificado de vasos de pressão de parede delgada.

Entretanto, o atendimento a esse critério geométrico não representa, isoladamente, uma garantia de segurança mecânica nem substitui as verificações exigidas pelas normas de projeto de vasos de pressão.

### 3. Formação da tensão circunferencial na parede cilíndrica

Quando um fluido exerce pressão sobre a superfície interna de um casco cilíndrico, as forças resultantes dessa pressão tendem a expandir radialmente o equipamento.

A resistência oferecida pelo material da parede à expansão do casco origina uma tensão de tração na direção circunferencial, denominada tensão circunferencial ou tensão de aro (*hoop stress*).

Essa tensão será representada por:

$$
\sigma_{\theta}
$$

Para compreender sua origem física, considere um segmento cilíndrico de comprimento $L_c$, diâmetro interno $D_i$ e espessura uniforme $t$.

Imagine que esse segmento seja dividido por um plano longitudinal que passa pelo eixo central do cilindro.

A pressão interna tende a separar as duas metades do casco, enquanto as forças resistentes desenvolvidas nas duas seções da parede atuam no sentido de manter o equilíbrio mecânico.

A determinação da tensão circunferencial pode, portanto, ser realizada aplicando-se o princípio fundamental do equilíbrio de forças.

#### 3.1. Força resultante produzida pela pressão

A pressão representa uma força distribuída por unidade de área:

$$
P=\frac{F}{A}
$$

Entretanto, como a pressão atua perpendicularmente à superfície curva do cilindro, a determinação da força resultante na direção de separação das duas metades deve considerar a área projetada da superfície interna.

Para um segmento cilíndrico de diâmetro interno $D_i$ e comprimento $L_c$, essa área projetada corresponde a:

$$
A_{\mathrm{proj}}=D_iL_c
$$

Portanto, a força resultante produzida pela diferença de pressão será:

$$
F_P=PA_{\mathrm{proj}}
$$

Substituindo a expressão da área projetada:

$$
\boxed{
F_P=PD_iL_c
}
$$

Essa força representa a tendência de separação das duas metades do casco cilíndrico.

#### 3.2. Forças resistentes desenvolvidas na parede

A tensão mecânica normal pode ser definida como a razão entre a força resistente e a área da seção transversal do material:

$$
\sigma=\frac{F}{A}
$$

Consequentemente:

$$
F=\sigma A
$$

Cada uma das duas seções longitudinais da parede cilíndrica apresenta uma área resistente correspondente a:

$$
A_{\mathrm{res}}=tL_c
$$

Considerando a aproximação de tensão uniforme ao longo da espessura, a força resistente desenvolvida em cada seção será:

$$
F_{\mathrm{res},1}=\sigma_{\theta}tL_c
$$

Como existem duas seções resistentes:

$$
F_{\mathrm{res}}=2\sigma_{\theta}tL_c
$$

Portanto:

$$
\boxed{
F_{\mathrm{res}}=2\sigma_{\theta}tL_c
}
$$

#### 3.3. Aplicação do equilíbrio de forças

Para que o segmento cilíndrico permaneça em equilíbrio mecânico, a força resultante produzida pela pressão deverá ser equilibrada pelas forças resistentes desenvolvidas na parede.

Assim:

$$
F_P=F_{\mathrm{res}}
$$

Substituindo as expressões anteriormente desenvolvidas:

$$
PD_iL_c=2\sigma_{\theta}tL_c
$$

Como o comprimento $L_c$ aparece em ambos os membros da equação, podemos simplificar:

$$
PD_i=2\sigma_{\theta}t
$$

Isolando a tensão circunferencial:

$$
\boxed{
\sigma_{\theta}=\frac{PD_i}{2t}
}
$$

Essa expressão constitui a equação fundamental utilizada no dimensionamento mecânico preliminar deste estudo.

Observe que a equação foi obtida diretamente da aplicação do equilíbrio de forças, considerando as hipóteses estabelecidas para o modelo de parede delgada.

Portanto, não se trata de uma expressão matemática isolada, mas de uma representação do comportamento mecânico do equipamento submetido à pressão interna.

#### 3.4. Interpretação física da tensão circunferencial

A equação:

$$
\sigma_{\theta}=\frac{PD_i}{2t}
$$

permite compreender a influência dos principais parâmetros geométricos e operacionais sobre o comportamento mecânico da parede cilíndrica.

Mantendo constantes as demais variáveis, podemos estabelecer as seguintes relações:

| Parâmetro | Relação com a tensão circunferencial |
|:---|:---|
| Pressão de projeto ($P$) | Diretamente proporcional |
| Diâmetro interno ($D_i$) | Diretamente proporcional |
| Espessura da parede ($t$) | Inversamente proporcional |
| Comprimento cilíndrico ($L_c$) | Não aparece na expressão obtida |

O aumento da pressão interna produz um aumento da força resultante que tende a separar as duas metades do cilindro.

Da mesma forma, o aumento do diâmetro interno amplia a área projetada sobre a qual a pressão atua, aumentando a força resultante.

Por outro lado, o aumento da espessura da parede amplia a área resistente do material, proporcionando uma redução da tensão circunferencial.

É particularmente interessante observar que o comprimento do casco não aparece na expressão final.

Isso ocorre porque o aumento do comprimento provoca simultaneamente um aumento da força produzida pela pressão e um aumento da área resistente da parede, mantendo a mesma relação entre essas grandezas no modelo simplificado.

Entretanto, o comprimento do casco influencia diretamente o volume de material, a massa e o custo de matéria-prima.

Portanto, embora o comprimento não determine a tensão circunferencial neste modelo, ele continua sendo um parâmetro importante para a avaliação física e econômica do equipamento.

#### 3.1. Seleção automática das espessuras comerciais

Após calcular a espessura mínima, precisamos identificar quais espessuras comerciais satisfazem a condição:

$$
t\geq t_{\min}
$$

Em Python, podemos realizar esse procedimento utilizando uma compreensão de lista (*list comprehension*).

Por exemplo, para uma espessura mínima de 7,50 mm, o conjunto original de espessuras comerciais é:

$$
\mathcal{T}=\{5,\ 6,\ 8,\ 10,\ 12,\ 16\}
$$

Após aplicar a restrição mecânica, obtemos o subconjunto:

$$
\mathcal{T}_{\mathrm{adequadas}}=
\{8,\ 10,\ 12,\ 16\}
$$

A menor espessura pertencente a esse subconjunto será selecionada:

$$
t_{\mathrm{adot}}=
\min\{8,\ 10,\ 12,\ 16\}
$$

$$
t_{\mathrm{adot}}=8\ \mathrm{mm}
$$

No programa, utilizaremos a função `min()` para identificar o menor elemento da lista de espessuras adequadas.

Também será necessário verificar se essa lista possui pelo menos um elemento.

Caso nenhuma espessura comercial satisfaça a condição estabelecida, o material não poderá ser dimensionado utilizando as espessuras disponíveis e deverá ser excluído da análise.

Esse tratamento permite que o programa continue avaliando os demais materiais candidatos.


````python
# Exemplo de selecao da espessura comercial

t_min_exemplo = 7.50

opcoes_exemplo = [
    t for t in espessuras_mm
    if t >= t_min_exemplo
]

print("Espessuras comerciais disponiveis:")
print(espessuras_mm)

print("\nEspessura minima calculada:")
print(f"{t_min_exemplo:.2f} mm")

print("\nEspessuras comerciais adequadas:")
print(opcoes_exemplo)

if opcoes_exemplo:
    t_adot_exemplo = min(opcoes_exemplo)

    print("\nEspessura comercial selecionada:")
    print(f"{t_adot_exemplo:.2f} mm")

else:
    print("\nNenhuma espessura comercial adequada.")
````

#### 3.2. Dimensionamento automatizado dos materiais candidatos

Uma vez compreendido o procedimento de seleção das espessuras comerciais, podemos incorporá-lo ao algoritmo de dimensionamento dos três materiais.

Utilizaremos a estrutura de repetição `for` para acessar as propriedades de cada material, calcular sua espessura mínima e selecionar a espessura comercial correspondente.

Durante as comparações numéricas, utilizaremos a tolerância `tol`, definida anteriormente, para evitar que pequenas imprecisões decorrentes da representação computacional de números reais interfiram na seleção de uma espessura.

Essa tolerância possui finalidade exclusivamente computacional e não representa uma tolerância de fabricação ou uma margem de segurança mecânica.

Os resultados serão armazenados sem arredondamentos intermediários. A formatação com duas casas decimais será utilizada exclusivamente na apresentação dos resultados.


````python
# ==================================================
# ETAPA 1 - DIMENSIONAMENTO PRELIMINAR
# ==================================================

# Dicionario para armazenamento dos resultados

resultados = {}

# Materiais sem espessura comercial adequada

sem_espessura = []

# Dimensionamento dos materiais candidatos

for nome, dados in materiais.items():

    # Tensao admissivel do material (MPa)

    S_MPa = dados["S_MPa"]

    # Calculo da espessura minima (mm)

    t_min_mm = P_MPa * D_mm / (2.0 * S_MPa)

    # Identificacao das espessuras comerciais adequadas

    opcoes = [
        t for t in espessuras_mm
        if t >= t_min_mm - tol
    ]

    # Verificacao da disponibilidade comercial

    if not opcoes:

        sem_espessura.append(nome)

        print(f"Material {nome}: sem espessura adequada.")

        continue

    # Selecao da menor espessura comercial adequada

    t_mm = min(opcoes)

    # Diferenca entre a espessura adotada e a minima

    delta_t_mm = t_mm - t_min_mm

    # Armazenamento dos resultados

    resultados[nome] = {
        "S_MPa": S_MPa,
        "t_min_mm": t_min_mm,
        "t_mm": t_mm,
        "delta_t_mm": delta_t_mm
    }
````

#### 3.3. Compreendendo o armazenamento dos resultados

Observe que, ao final de cada repetição, os resultados calculados são armazenados em um dicionário.

Para cada material, são armazenadas as seguintes grandezas:

| Chave | Grandeza armazenada | Unidade |
|:---|:---|:---:|
| `S_MPa` | Tensão admissível | MPa |
| `t_min_mm` | Espessura mínima calculada | mm |
| `t_mm` | Espessura comercial adotada | mm |
| `delta_t_mm` | Diferença entre a espessura adotada e a mínima | mm |

A diferença entre a espessura comercial adotada e a espessura mínima é calculada por:

$$
\Delta t=t_{\mathrm{adot}}-t_{\min}
$$

Essa grandeza permite identificar quanto a espessura comercial selecionada excede a espessura mínima teórica.

Não devemos confundir essa diferença com uma sobre-espessura de corrosão, uma tolerância de fabricação ou um fator de segurança.

Trata-se exclusivamente da diferença resultante da seleção de uma espessura comercial.

### 6. Implementação computacional da verificação mecânica e dos cálculos geométricos

Na etapa anterior, armazenamos as espessuras mínimas e comerciais de cada material no dicionário `resultados`.

Nesta etapa, utilizaremos essas informações para calcular as novas grandezas, seguindo a sequência:

1. Recuperar a espessura comercial adotada;
2. Calcular a tensão circunferencial;
3. Verificar o atendimento ao critério mecânico;
4. Calcular as folgas absoluta e percentual de tensão;
5. Converter a espessura para metros;
6. Determinar o diâmetro externo;
7. Calcular o volume de material;
8. Recuperar a massa específica do material;
9. Calcular a massa da parede cilíndrica;
10. Acrescentar os novos resultados ao dicionário.

A utilização de uma estrutura de dados integrada permite desenvolver o programa progressivamente, preservando os resultados obtidos nas etapas anteriores.

Esse procedimento evita a repetição desnecessária de cálculos e facilita a organização da memória de cálculo computacional.

````python
# ==================================================
# ETAPA 2.1 - VERIFICACAO MECANICA
# ==================================================

for nome, r in resultados.items():

    # Recuperacao dos resultados da Etapa 1

    t_mm = r["t_mm"]
    S_MPa = r["S_MPa"]

    # Tensao circunferencial (MPa)

    sigma_MPa = P_MPa * D_mm / (2.0 * t_mm)

    # Verificacao do criterio mecanico

    atende_mecanica = (
        sigma_MPa <= S_MPa + tol
    )

    # Folga absoluta de tensao (MPa)

    folga_MPa = S_MPa - sigma_MPa

    # Folga percentual de tensao (%)

    folga_percentual = (
        folga_MPa / S_MPa
    ) * 100.0

    # Armazenamento dos novos resultados

    r["sigma_MPa"] = sigma_MPa
    r["mecanica"] = atende_mecanica
    r["folga_MPa"] = folga_MPa
    r["folga_percentual"] = folga_percentual
````

### 7. Implementação computacional dos cálculos de volume e massa

Uma vez realizada a verificação mecânica, podemos utilizar as espessuras comerciais adotadas para determinar as características geométricas e físicas da parede cilíndrica.

Inicialmente, converteremos a espessura de milímetros para metros:

$$
t_{\mathrm{m}}=\frac{t_{\mathrm{mm}}}{1000}
$$

Em seguida, determinaremos o diâmetro externo:

$$
D_e=D_i+2t
$$

O volume de material será calculado utilizando a expressão geométrica exata:

$$
V_{\mathrm{mat}}=\pi L_c(D_it+t^2)
$$

Finalmente, recuperaremos a massa específica de cada material, armazenada no dicionário `materiais`, e calcularemos a massa correspondente:

$$
m=\rho V_{\mathrm{mat}}
$$

Observe que, nesta etapa, utilizaremos simultaneamente dois dicionários:

- `materiais`, contendo as propriedades originais dos materiais;
- `resultados`, contendo as grandezas determinadas durante a resolução do problema.

Essa integração entre os dados de entrada e os resultados calculados constitui uma característica importante da organização do programa computacional.

````python
# ==================================================
# ETAPA 2.2 - CALCULO DOS VOLUMES E DAS MASSAS
# ==================================================

for nome, r in resultados.items():

    # Conversao da espessura para metros

    t_m = r["t_mm"] / 1000.0

    # Diametro externo (m)

    D_ext_m = D_m + 2.0 * t_m

    # Volume de material da parede cilindrica (m3)

    volume_m3 = math.pi * L_m * (
        D_m * t_m + t_m**2
    )

    # Massa especifica do material (kg/m3)

    rho = materiais[nome]["rho"]

    # Massa da parede cilindrica (kg)

    massa_kg = rho * volume_m3

    # Armazenamento dos novos resultados

    r["t_m"] = t_m
    r["D_ext_m"] = D_ext_m
    r["volume_m3"] = volume_m3
    r["rho"] = rho
    r["massa_kg"] = massa_kg
````

````python
# ==================================================
# COMPARACAO QUANTITATIVA DOS VOLUMES E DAS MASSAS
# ==================================================

from itertools import combinations

# Comparacao de todos os pares de materiais

for referencia, alternativa in combinations(resultados, 2):

    r_ref = resultados[referencia]
    r_alt = resultados[alternativa]

    # Diferenca absoluta de volume

    delta_volume = (
        r_ref["volume_m3"] - r_alt["volume_m3"]
    )

    # Reducao percentual de volume

    reducao_volume = (
        delta_volume / r_ref["volume_m3"]
    ) * 100.0

    # Diferenca absoluta de massa

    delta_massa = (
        r_ref["massa_kg"] - r_alt["massa_kg"]
    )

    # Reducao percentual de massa

    reducao_massa = (
        delta_massa / r_ref["massa_kg"]
    ) * 100.0

    # Apresentacao dos resultados

    print(f"\nMATERIAL {alternativa} EM RELACAO AO "
          f"MATERIAL {referencia}")

    print("-" * 55)

    print(
        f"Diferenca de volume: {delta_volume:.6f} m3"
    )

    print(
        f"Reducao percentual de volume: "
        f"{reducao_volume:.2f}%"
    )

    print(
        f"Diferenca de massa: {delta_massa:.2f} kg"
    )

    print(
        f"Reducao percentual de massa: "
        f"{reducao_massa:.2f}%"
    )
````

Para realizar as comparações entre os materiais, utilizamos a função `combinations()`, pertencente ao módulo `itertools` da biblioteca padrão do Python.

Essa função permite gerar todas as combinações possíveis de dois elementos, sem repetir os pares.

Para os três materiais candidatos, serão produzidas as seguintes combinações:

Em cada comparação, o primeiro material será utilizado como referência para o cálculo das diferenças percentuais.

É importante compreender que o resultado de uma comparação percentual depende da grandeza utilizada como referência.

Por exemplo, afirmar que o Material C apresenta uma massa 17,78% inferior à do Material B não significa que o Material B apresenta uma massa 17,78% superior à do Material C.

As bases de referência são diferentes.

No algoritmo desenvolvido, uma redução percentual positiva indica que a alternativa analisada apresenta um valor inferior ao material de referência.

Caso o resultado seja negativo, isso indicará um aumento, e não uma redução.

Essa convenção deverá ser observada durante a interpretação dos resultados computacionais.

A identificação explícita da base de referência constitui uma boa prática na elaboração de relatórios técnicos de engenharia, evitando ambiguidades na comunicação dos resultados.

### 4. Implementação computacional da avaliação econômica

Para automatizar o cálculo dos custos de matéria-prima, utilizaremos novamente uma estrutura de repetição para percorrer os materiais dimensionados nas etapas anteriores.

O algoritmo será desenvolvido de acordo com a seguinte sequência:

1. Identificar o material analisado;
2. Recuperar a massa calculada na Etapa 2;
3. Recuperar o custo unitário armazenado nos dados de entrada;
4. Calcular o custo de matéria-prima;
5. Armazenar o custo unitário e o custo calculado no dicionário de resultados;
6. Repetir o procedimento para os demais materiais.

Observe que não será necessário calcular novamente as espessuras, os volumes ou as massas.

Essas informações já foram determinadas e armazenadas nas etapas anteriores.

Essa característica demonstra uma das vantagens da organização progressiva de uma memória de cálculo computacional: cada etapa utiliza os resultados produzidos anteriormente, evitando a repetição desnecessária de operações.

````python
# ==================================================
# ETAPA 3 - CALCULO DOS CUSTOS DE MATERIA-PRIMA
# ==================================================

for nome, r in resultados.items():

    # Recuperacao da massa calculada na Etapa 2

    massa_kg = r["massa_kg"]

    # Recuperacao do custo unitario (R$/kg)

    preco = materiais[nome]["preco"]

    # Calculo do custo de materia-prima (R$)

    custo_R = massa_kg * preco

    # Armazenamento dos novos resultados

    r["preco"] = preco
    r["custo_R"] = custo_R
````

| `preco` | Custo unitário da matéria-prima | R$/kg |
| `custo_R` | Custo estimado de matéria-prima | R$ |

Observe que a instrução:

```python
custo_R = massa_kg * preco
```

representa diretamente a equação matemática:

$$
C=mc_{\mathrm{kg}}
$$

A implementação computacional torna-se simples quando o modelo matemático e as estruturas de dados foram previamente estabelecidos.

É importante destacar que nenhum arredondamento foi realizado durante o cálculo dos custos.

Os valores serão armazenados com a precisão numérica disponível, sendo arredondados para duas casas decimais apenas durante sua apresentação.

Essa prática evita a propagação desnecessária de erros de arredondamento ao longo da memória de cálculo.

### 5. Classificação econômica dos materiais candidatos

Após calcular os custos de matéria-prima, podemos organizar os materiais em ordem crescente de custo.

Essa classificação permite identificar:

- A alternativa de menor custo;
- As alternativas de custo intermediário;
- A alternativa de maior custo.

Em Python, podemos utilizar a função `sorted()` para ordenar os elementos de uma estrutura de dados segundo determinado critério.

Neste problema, o critério de ordenação será o custo de matéria-prima armazenado no dicionário `resultados`.

Utilizaremos a expressão:

```python
sorted(resultados, key=lambda nome: resultados[nome]["custo_R"])
```

A função `sorted()` produzirá uma lista contendo os nomes dos materiais organizados em ordem crescente de custo.

O argumento `key` estabelece qual informação será utilizada como critério de ordenação.

A expressão `lambda` representa uma função anônima que, para cada material, retorna o custo armazenado no respectivo dicionário de resultados.

Dessa maneira, a classificação será determinada automaticamente pelos resultados calculados, e não por uma ordem previamente estabelecida pelo programador.

É importante compreender que essa classificação considera exclusivamente o custo de matéria-prima, sem aplicar ainda as restrições de admissibilidade técnica.

````python
# ==================================================
# CLASSIFICACAO ECONOMICA DOS MATERIAIS
# ==================================================

# Ordenacao dos materiais por custo crescente

ordem_economica = sorted(
    resultados,
    key=lambda nome: resultados[nome]["custo_R"]
)

print("CLASSIFICACAO ECONOMICA PRELIMINAR")
print("=" * 50)

for posicao, nome in enumerate(ordem_economica, start=1):

    custo = resultados[nome]["custo_R"]

    print(
        f"{posicao} - Material {nome}: "
        f"R$ {custo:.2f}"
    )

print("=" * 50)

# Identificacao do material de menor custo

if ordem_economica:

    menor_custo = ordem_economica[0]

    print(
        f"\nMaterial de menor custo: {menor_custo}"
    )

    print(
        f"Custo de materia-prima: "
        f"R$ {resultados[menor_custo]['custo_R']:.2f}"
    )

else:

    print("\nNenhum material foi dimensionado.")
````

### 7. Análise quantitativa das diferenças econômicas

Para compreender a magnitude das diferenças econômicas entre os materiais, calcularemos as diferenças absolutas e percentuais dos custos de matéria-prima.

A diferença absoluta entre o custo de uma alternativa e o custo de um material de referência será determinada por:

$$
\Delta C=C_{\mathrm{alt}}-C_{\mathrm{ref}}
$$

A diferença percentual será determinada por:

$$
\boxed{
R_C=
\frac{C_{\mathrm{alt}}-C_{\mathrm{ref}}}
{C_{\mathrm{ref}}}\times100
}
$$

Um resultado positivo indica que a alternativa analisada apresenta um custo superior ao material de referência.

Um resultado negativo indica que a alternativa analisada apresenta um custo inferior.

Nos cálculos seguintes, serão utilizados os valores não arredondados dos custos. Pequenas diferenças de centavos podem surgir caso sejam utilizados diretamente os valores arredondados apresentados nas tabelas.

$$
\boxed{
\Delta C_{B-A}\approx\mathrm{R\$}\ 5.899,25
}
$$

Tomando o Material A como referência:

$$
R_C=
\frac{C_B-C_A}{C_A}\times100
$$

Obtemos:

$$
\boxed{R_C\approx128,92\%}
$$

Portanto, o custo de matéria-prima do Material B é aproximadamente 128,92% superior ao custo do Material A.

Essa diferença é significativa do ponto de vista econômico.

#### 7.4. Economia proporcionada pelo Material B em relação ao Material C

Outra maneira de interpretar a diferença econômica entre os Materiais B e C consiste em determinar a economia proporcionada pela utilização do Material B, tomando o custo do Material C como referência.

A economia absoluta corresponde a:

$$
E_C=C_C-C_B
$$

Portanto:

$$
\boxed{
E_C\approx\mathrm{R\$}\ 2.085,52
}
$$

A economia percentual será determinada por:

$$
E_{\%}=
\frac{C_C-C_B}{C_C}\times100
$$

Obtemos:

$$
\boxed{E_{\%}\approx16,60\%}
$$

Portanto, a utilização do Material B em substituição ao Material C proporciona uma economia de aproximadamente R$ 2.085,52, correspondente a 16,60% do custo de matéria-prima do Material C.

Observe que os resultados de 19,91% e 16,60% não são contraditórios.

O primeiro representa o aumento percentual do custo do Material C em relação ao Material B.

O segundo representa a redução percentual do custo do Material B em relação ao Material C.

A diferença decorre exclusivamente da utilização de diferentes bases de referência.

Essa distinção deverá ser cuidadosamente observada durante a elaboração do relatório técnico de engenharia.

````python
# ==================================================
# COMPARACAO QUANTITATIVA DOS CUSTOS
# ==================================================

for referencia, alternativa in combinations(resultados, 2):

    # Recuperacao dos custos

    custo_ref = resultados[referencia]["custo_R"]
    custo_alt = resultados[alternativa]["custo_R"]

    # Diferenca absoluta de custo

    delta_custo = custo_alt - custo_ref

    # Variacao percentual em relacao a referencia

    variacao_percentual = (
        delta_custo / custo_ref
    ) * 100.0

    # Apresentacao dos resultados

    print(
        f"\nMATERIAL {alternativa} EM RELACAO AO "
        f"MATERIAL {referencia}"
    )

    print("-" * 60)

    print(
        f"Custo de referencia: R$ {custo_ref:.2f}"
    )

    print(
        f"Custo da alternativa: R$ {custo_alt:.2f}"
    )

    print(
        f"Diferenca de custo: R$ {delta_custo:.2f}"
    )

    print(
        f"Variacao percentual: "
        f"{variacao_percentual:.2f}%"
    )
````

### 8. Influência da massa e do custo unitário sobre o custo de matéria-prima

Os resultados obtidos demonstram que o Material C proporciona a construção da parede cilíndrica de menor massa.

Conforme determinado na Etapa 2, a utilização do Material C em relação ao Material B proporciona uma redução de massa correspondente a:

$$
R_m=
\frac{m_B-m_C}{m_B}\times100
$$

$$
\boxed{R_m\approx17,78\%}
$$

Entretanto, o custo unitário do Material C é superior ao custo unitário do Material B.

Esse aumento percentual pode ser determinado por:

$$
R_p=
\frac{c_{\mathrm{kg},C}-c_{\mathrm{kg},B}}
{c_{\mathrm{kg},B}}\times100
$$

Substituindo os valores:

$$
R_p=\frac{35-24}{24}\times100
$$

Portanto:

$$
\boxed{R_p\approx45,83\%}
$$

Temos, portanto, dois efeitos simultâneos:

- Redução de aproximadamente 17,78% na massa de matéria-prima;
- Aumento de aproximadamente 45,83% no custo unitário.

Para compreender o efeito combinado dessas duas alterações, podemos estabelecer a razão entre os custos dos Materiais C e B:

$$
\frac{C_C}{C_B}
=
\frac{m_Cc_{\mathrm{kg},C}}
{m_Bc_{\mathrm{kg},B}}
$$

Reorganizando a expressão:

$$
\boxed{
\frac{C_C}{C_B}
=
\frac{m_C}{m_B}
\frac{c_{\mathrm{kg},C}}{c_{\mathrm{kg},B}}
}
$$

Substituindo os valores calculados:

$$
\frac{C_C}{C_B}
\approx
\frac{358,876695190}{436,465237274}
\frac{35}{24}
$$

$$
\frac{C_C}{C_B}
\approx0,822234\times1,458333
$$

Consequentemente:

$$
\frac{C_C}{C_B}\approx1,199092
$$

Esse resultado demonstra que o custo do Material C corresponde a aproximadamente 119,91% do custo do Material B.

Portanto, o aumento percentual do custo será:

$$
R_C=
\left(
\frac{C_C}{C_B}-1
\right)\times100
$$

$$
\boxed{R_C\approx19,91\%}
$$

Concluímos que a redução de massa proporcionada pelo Material C não foi suficiente para compensar seu maior custo unitário.

Observe que os efeitos percentuais da massa e do custo unitário não devem ser simplesmente somados ou subtraídos.

Como o custo total resulta do produto entre essas duas grandezas, seus efeitos combinados também devem ser avaliados de maneira multiplicativa.

Essa análise demonstra a importância de compreender as relações matemáticas existentes entre as variáveis de um problema de engenharia antes de interpretar os resultados computacionais.

````python
# ==================================================
# INFLUENCIA DA MASSA E DO CUSTO UNITARIO
# ==================================================

if "B" in resultados and "C" in resultados:

    massa_B = resultados["B"]["massa_kg"]
    massa_C = resultados["C"]["massa_kg"]

    preco_B = resultados["B"]["preco"]
    preco_C = resultados["C"]["preco"]

    # Razao entre as massas

    razao_massa = massa_C / massa_B

    # Razao entre os custos unitarios

    razao_preco = preco_C / preco_B

    # Razao entre os custos totais

    razao_custo = razao_massa * razao_preco

    # Variacoes percentuais

    reducao_massa = (1.0 - razao_massa) * 100.0

    aumento_preco = (razao_preco - 1.0) * 100.0

    aumento_custo = (razao_custo - 1.0) * 100.0

    print("MATERIAL C EM RELACAO AO MATERIAL B")
    print("=" * 60)

    print(f"Reducao da massa: {reducao_massa:.2f}%")

    print(f"Aumento do custo unitario: {aumento_preco:.2f}%")

    print(f"Aumento do custo total: {aumento_custo:.2f}%")

    print("-" * 60)

    print(f"Razao entre as massas: {razao_massa:.6f}")

    print(f"Razao entre os precos: {razao_preco:.6f}")

    print(f"Razao entre os custos: {razao_custo:.6f}")

    print("=" * 60)

else:

    print("Comparacao indisponivel para B e C.")
````

### 9. Determinação do custo unitário de equilíbrio econômico

Até este momento, verificamos que o Material C apresenta uma massa inferior à do Material B, mas seu custo unitário é suficientemente elevado para tornar seu custo total de matéria-prima superior.

Entretanto, os preços dos materiais podem sofrer alterações.

Podemos, então, formular a seguinte questão:

> Qual deveria ser o custo unitário do Material C para que seu custo total de matéria-prima fosse igual ao custo do Material B?

Para responder a essa questão, estabeleceremos a condição de equilíbrio econômico:

$$
C_C=C_B
$$

Substituindo as expressões dos custos:

$$
m_Cc_{\mathrm{kg},C}^{*}
=
m_Bc_{\mathrm{kg},B}
$$

Isolando o custo unitário do Material C:

$$
\boxed{
c_{\mathrm{kg},C}^{*}
=
\frac{m_Bc_{\mathrm{kg},B}}{m_C}
}
$$

Como:

$$
C_B=m_Bc_{\mathrm{kg},B}
$$

podemos escrever:

$$
\boxed{
c_{\mathrm{kg},C}^{*}=\frac{C_B}{m_C}
}
$$

Substituindo os valores calculados:

$$
c_{\mathrm{kg},C}^{*}
\approx
\frac{10475,165694586}{358,876695190}
$$

Obtemos:

$$
\boxed{
c_{\mathrm{kg},C}^{*}
\approx29,19\ \mathrm{R\$/kg}
}
$$

Portanto, mantendo constantes as demais condições do problema, o Material C deverá apresentar um custo unitário de aproximadamente R$ 29,19/kg para que seu custo de matéria-prima seja aproximadamente igual ao custo do Material B.

Acima desse valor, o Material B será economicamente mais vantajoso.

Abaixo do valor exato de equilíbrio, o Material C apresentará menor custo de matéria-prima.

Essa análise demonstra que uma decisão de engenharia pode ser modificada quando determinadas condições econômicas são alteradas.

Por esse motivo, a seleção de materiais deve estar fundamentada em um modelo matemático e computacional que permita investigar diferentes cenários, e não apenas reproduzir uma solução numérica para um conjunto fixo de dados.

````python
# ==================================================
# CUSTO UNITARIO DE EQUILIBRIO ECONOMICO
# ==================================================

if "B" in resultados and "C" in resultados:

    # Custo de materia-prima do Material B

    custo_B = resultados["B"]["custo_R"]

    # Massa do Material C

    massa_C = resultados["C"]["massa_kg"]

    # Preco atual do Material C

    preco_C = resultados["C"]["preco"]

    # Preco de equilibrio economico

    preco_equilibrio = custo_B / massa_C

    # Reducao necessaria do preco unitario

    reducao_preco = preco_C - preco_equilibrio

    # Reducao percentual

    reducao_percentual = (
        reducao_preco / preco_C
    ) * 100.0

    print("EQUILIBRIO ECONOMICO ENTRE B E C")
    print("=" * 60)

    print(
        f"Preco atual do Material C: "
        f"R$ {preco_C:.2f}/kg"
    )

    print(
        f"Preco de equilibrio: "
        f"R$ {preco_equilibrio:.2f}/kg"
    )

    print(
        f"Reducao necessaria: "
        f"R$ {reducao_preco:.2f}/kg"
    )

    print(
        f"Reducao percentual: "
        f"{reducao_percentual:.2f}%"
    )

    print("=" * 60)

else:

    print("Comparacao indisponivel para B e C.")
````

### 2. Avaliação das restrições técnicas

Em um problema de seleção de materiais, determinadas condições não podem ser tratadas simplesmente como vantagens ou desvantagens de uma alternativa.

Algumas condições representam requisitos obrigatórios de projeto.

Neste estudo, consideraremos três requisitos fundamentais:

1. Existência de uma espessura comercial adequada;
2. Atendimento ao critério mecânico simplificado;
3. Compatibilidade química com o fluido de processo.

Um material somente será considerado tecnicamente admissível quando satisfizer simultaneamente esses três requisitos.

#### 2.1. Disponibilidade de espessura comercial

A espessura comercial selecionada deverá satisfazer a condição:

$$
t_{\mathrm{adot}}\geq t_{\min}
$$

Caso nenhuma das espessuras disponíveis satisfaça essa condição, o material não poderá ser selecionado nas condições estabelecidas.

#### 2.2. Atendimento ao critério mecânico

A tensão circunferencial desenvolvida na parede cilíndrica deverá satisfazer a condição:

$$
\sigma_{\theta}\leq S_{\mathrm{adm}}
$$

Caso a tensão calculada ultrapasse a tensão admissível, o material não atenderá ao critério mecânico simplificado.

#### 2.3. Compatibilidade química

O material deverá ser compatível com o fluido de processo.

A incompatibilidade química constitui uma restrição eliminatória neste problema, independentemente do custo, da massa ou da resistência mecânica do material.

Em um equipamento industrial real, a utilização de materiais incompatíveis com o meio de operação pode provocar mecanismos de degradação capazes de comprometer sua integridade e sua vida útil.

Neste exercício, a compatibilidade química é uma informação previamente fornecida nos dados de entrada, não sendo necessário desenvolver um modelo adicional de degradação química.

#### 2.4. Admissibilidade técnica

Considerando os três requisitos, podemos representar a condição de admissibilidade por:

$$
\boxed{
\mathrm{Admissibilidade}
=
E\land M\land Q
}
$$

em que:

- $E$ representa a existência de uma espessura comercial adequada;
- $M$ representa o atendimento ao critério mecânico;
- $Q$ representa a compatibilidade química.

O operador lógico $\land$ representa a operação lógica E, denominada `and` na linguagem Python.

Portanto, a admissibilidade técnica somente será verdadeira quando todas as condições forem simultaneamente verdadeiras.

### 5. Implementação computacional da seleção de materiais

O algoritmo de seleção será desenvolvido utilizando os resultados armazenados durante as etapas anteriores.

Inicialmente, verificaremos a compatibilidade química de cada material e recuperaremos o resultado da verificação mecânica.

A existência de uma espessura comercial adequada já foi verificada durante a Etapa 1.

Os materiais para os quais não foi possível selecionar uma espessura comercial foram registrados na lista `sem_espessura` e não foram incorporados ao dicionário `resultados`.

Portanto, para os materiais presentes nesse dicionário, a admissibilidade será determinada pela expressão lógica:

```python
admissivel = atende_mecanica and compativel
```

Essa expressão somente retornará o valor `True` quando ambas as condições forem verdadeiras.

Após determinar a admissibilidade de cada material, construiremos um novo dicionário contendo exclusivamente as alternativas tecnicamente admissíveis.

Finalmente, utilizaremos a função `min()` para identificar o material de menor custo entre essas alternativas.

O procedimento deverá distinguir claramente dois resultados:

- Material de menor custo entre todas as alternativas dimensionadas;
- Material de menor custo entre as alternativas tecnicamente admissíveis.

Essa distinção é fundamental para que a decisão computacional represente corretamente o problema de engenharia formulado.

````python
# ==================================================
# ETAPA 4.1 - AVALIACAO DAS RESTRICOES TECNICAS
# ==================================================

for nome, r in resultados.items():

    # Recuperacao da compatibilidade quimica

    compativel = materiais[nome]["compativel"]

    # Recuperacao da verificacao mecanica

    atende_mecanica = r["mecanica"]

    # Avaliacao da admissibilidade tecnica

    admissivel = atende_mecanica and compativel

    # Armazenamento dos resultados

    r["compativel"] = compativel
    r["admissivel"] = admissivel

    # Identificacao dos motivos de exclusao

    motivos = []

    if not atende_mecanica:

        motivos.append(
            "Nao atende ao criterio mecanico"
        )

    if not compativel:

        motivos.append(
            "Incompatibilidade quimica"
        )

    # Registro da justificativa

    if admissivel:

        r["justificativa"] = (
            "Atende aos criterios tecnicos"
        )

    else:

        r["justificativa"] = "; ".join(motivos)
````

#### 5.1. Utilização de operadores lógicos na tomada de decisão

A instrução:

```python
admissivel = atende_mecanica and compativel
```

representa computacionalmente a exigência de atendimento simultâneo aos critérios mecânico e químico.

O comportamento do operador lógico `and` pode ser compreendido por meio da seguinte tabela-verdade.

| Critério mecânico | Compatibilidade química | Admissibilidade |
|:---:|:---:|:---:|
| True | True | True |
| True | False | False |
| False | True | False |
| False | False | False |

Observe que o atendimento a apenas um dos critérios não é suficiente para tornar o material admissível.

Essa característica representa corretamente a natureza das restrições estabelecidas para o problema.

Também utilizamos uma lista denominada `motivos` para armazenar as razões pelas quais determinado material foi considerado não admissível.

Essa estratégia permite que o programa apresente não apenas uma classificação, mas também uma justificativa para a exclusão de uma alternativa.

A apresentação dessas justificativas contribui para a transparência do processo computacional de tomada de decisão.

````python
# ==================================================
# APRESENTACAO DA AVALIACAO TECNICA
# ==================================================

print("AVALIACAO DAS RESTRICOES TECNICAS")
print("=" * 65)

for nome, r in resultados.items():

    print(f"\nMATERIAL {nome}")

    print("-" * 65)

    mecanica = (
        "Atende" if r["mecanica"]
        else "Nao atende"
    )

    quimica = (
        "Compativel" if r["compativel"]
        else "Incompativel"
    )

    situacao = (
        "Admissivel" if r["admissivel"]
        else "Nao admissivel"
    )

    print(f"Criterio mecanico: {mecanica}")
    print(f"Compatibilidade quimica: {quimica}")
    print(f"Situacao tecnica: {situacao}")

    print(
        f"Justificativa: {r['justificativa']}"
    )

# Materiais excluidos na Etapa 1

for nome in sem_espessura:

    print(f"\nMATERIAL {nome}")

    print("-" * 65)

    print("Situacao tecnica: Nao admissivel")

    print(
        "Justificativa: nenhuma espessura "
        "comercial adequada."
    )

print("\n" + "=" * 65)
````

### 6. Construção do conjunto de materiais admissíveis

Após avaliar as restrições técnicas, precisamos separar as alternativas admissíveis daquelas que não poderão ser selecionadas.

Matematicamente, esse procedimento corresponde à construção de um subconjunto do conjunto original de materiais candidatos.

Em Python, podemos realizar essa operação utilizando uma compreensão de dicionário (*dictionary comprehension*).

A expressão:

```python
admissiveis = {
    nome: r
    for nome, r in resultados.items()
    if r["admissivel"]
}
```

constrói um novo dicionário contendo exclusivamente os materiais cuja condição de admissibilidade seja verdadeira.

Observe que os materiais excluídos continuam armazenados no dicionário original `resultados`.

Essa característica é importante porque os resultados das alternativas não admissíveis também possuem valor para a análise de engenharia.

Por exemplo, embora o Material A não possa ser recomendado, seu custo de matéria-prima permite compreender quantitativamente as consequências econômicas da restrição de compatibilidade química.

Portanto, não devemos eliminar essas informações da memória de cálculo.

Devemos apenas impedir que as alternativas não admissíveis participem da seleção final.

````python
# ==================================================
# ETAPA 4.2 - CONJUNTO DE MATERIAIS ADMISSIVEIS
# ==================================================

admissiveis = {
    nome: r
    for nome, r in resultados.items()
    if r["admissivel"]
}

nao_admissiveis = [
    nome
    for nome, r in resultados.items()
    if not r["admissivel"]
]

# Inclusao dos materiais sem espessura adequada

nao_admissiveis.extend(sem_espessura)

# Apresentacao dos resultados

print("CLASSIFICACAO TECNICA DOS MATERIAIS")
print("=" * 55)

print(
    f"Numero de materiais candidatos: "
    f"{len(materiais)}"
)

print(
    f"Numero de materiais admissiveis: "
    f"{len(admissiveis)}"
)

print(
    f"Numero de materiais nao admissiveis: "
    f"{len(nao_admissiveis)}"
)

print(
    f"\nMateriais admissiveis: "
    f"{list(admissiveis.keys())}"
)

print(
    f"Materiais nao admissiveis: "
    f"{nao_admissiveis}"
)

print("=" * 55)
````

### 7. Seleção automática da alternativa de menor custo

Uma vez construído o conjunto de materiais admissíveis, podemos aplicar a função objetivo do problema.

Conforme estabelecido na fundamentação matemática:

$$
i^{*}=
\underset{i\in\mathcal{A}}{\operatorname{argmin}}\ C_i
$$

Na linguagem Python, podemos realizar essa operação utilizando a função `min()` associada a um critério de comparação.

Inicialmente, identificaremos o material de menor custo entre todas as alternativas dimensionadas:

```python
mais_barato = min(
    resultados,
    key=lambda nome: resultados[nome]["custo_R"]
)
```

Posteriormente, aplicaremos o mesmo procedimento exclusivamente ao conjunto de materiais admissíveis:

```python
recomendado = min(
    admissiveis,
    key=lambda nome: admissiveis[nome]["custo_R"]
)
```

Observe que as duas operações utilizam exatamente o mesmo critério econômico.

A diferença encontra-se no conjunto de alternativas submetido à comparação.

Na primeira operação, todos os materiais dimensionados participam da análise.

Na segunda operação, somente os materiais que satisfazem as restrições técnicas são considerados.

Essa diferença representa computacionalmente a aplicação das restrições ao problema de otimização.

Também será necessário verificar se os dicionários possuem elementos antes de aplicar a função `min()`.

Caso nenhum material seja admissível, o programa deverá informar que não existe uma solução viável nas condições estabelecidas, em vez de apresentar uma recomendação inadequada.

````python
# ==================================================
# ETAPA 4.3 - SELECAO AUTOMATICA DO MATERIAL
# ==================================================

# Inicializacao das variaveis de selecao

mais_barato = None
recomendado = None

# Material de menor custo entre os dimensionados

if resultados:

    mais_barato = min(
        resultados,
        key=lambda nome: resultados[nome]["custo_R"]
    )

# Material de menor custo entre os admissiveis

if admissiveis:

    recomendado = min(
        admissiveis,
        key=lambda nome: admissiveis[nome]["custo_R"]
    )

# Apresentacao dos resultados

print("SELECAO PRELIMINAR DO MATERIAL")
print("=" * 65)

if mais_barato is not None:

    r_barato = resultados[mais_barato]

    print("\nALTERNATIVA DE MENOR CUSTO")
    print("-" * 65)

    print(f"Material: {mais_barato}")

    print(
        f"Custo de materia-prima: "
        f"R$ {r_barato['custo_R']:.2f}"
    )

    print(
        f"Admissibilidade tecnica: "
        f"{'Sim' if r_barato['admissivel'] else 'Nao'}"
    )

    print(
        f"Justificativa: "
        f"{r_barato['justificativa']}"
    )

else:

    print("\nNenhum material foi dimensionado.")

if recomendado is not None:

    r_rec = admissiveis[recomendado]

    print("\nALTERNATIVA TECNICAMENTE RECOMENDADA")
    print("-" * 65)

    print(f"Material: {recomendado}")

    print(
        f"Custo de materia-prima: "
        f"R$ {r_rec['custo_R']:.2f}"
    )

    print("Admissibilidade tecnica: Sim")

    print(
        "Justificativa: menor custo entre "
        "os materiais admissiveis."
    )

else:

    print("\nNenhum material tecnicamente admissivel.")

    print(
        "Nao existe uma solucao viavel "
        "nas condicoes estabelecidas."
    )

print("\n" + "=" * 65)
````

### 10. Representação gráfica da seleção preliminar

Para representar graficamente a decisão de engenharia, construiremos um gráfico de barras contendo os custos de matéria-prima de todos os materiais dimensionados.

````python
# ==================================================
# REPRESENTACAO GRAFICA DA SELECAO FINAL
# ==================================================

from matplotlib.patches import Patch

if resultados:

    nomes = list(resultados.keys())

    custos = [
        resultados[n]["custo_R"]
        for n in nomes
    ]

    cores = []

    for nome in nomes:

        if not resultados[nome]["admissivel"]:

            cores.append("#A6A6A6")

        elif nome == recomendado:

            cores.append("#005A8D")

        else:

            cores.append("#8BC4E8")

    # Criacao do grafico

    fig, ax = plt.subplots(figsize=(10, 6))

    barras = ax.bar(
        [f"Material {n}" for n in nomes],
        custos,
        color=cores,
        width=0.50,
        edgecolor="black"
    )

    # Identificacao das barras

    for barra, nome, custo in zip(
        barras, nomes, custos
    ):

        if not resultados[nome]["admissivel"]:

            situacao = "Não admissível"

        elif nome == recomendado:

            situacao = "Recomendado"

        else:

            situacao = "Admissível"

        ax.annotate(
            f"R$ {custo:.2f}\n{situacao}",
            xy=(
                barra.get_x() + barra.get_width() / 2,
                barra.get_height()
            ),
            xytext=(0, 6),
            textcoords="offset points",
            ha="center",
            va="bottom",
            fontsize=10
        )

    # Legenda

    legenda = [
        Patch(
            facecolor="#A6A6A6",
            edgecolor="black",
            label="Não admissível"
        ),
        Patch(
            facecolor="#8BC4E8",
            edgecolor="black",
            label="Admissível"
        ),
        Patch(
            facecolor="#005A8D",
            edgecolor="black",
            label="Recomendado"
        )
    ]

    ax.legend(handles=legenda, loc="upper left")

    ax.set_xlabel("Materiais candidatos")

    ax.set_ylabel("Custo de matéria-prima (R$)")

    ax.set_title(
        "Seleção preliminar de materiais\n"
        "Minimização do custo com restrições técnicas"
    )

    ax.set_ylim(0, max(custos) * 1.45)

    ax.grid(axis="y", linestyle="--", alpha=0.35)

    ax.set_axisbelow(True)

    plt.tight_layout()
    plt.show()

else:

    print("Nenhum resultado disponivel para o grafico.")
````

Embora sua massa específica de 8000 kg/m³ seja superior à do Material A, a redução da espessura comercial proporcionou uma diminuição significativa da quantidade de matéria-prima, resultando em uma massa de aproximadamente 436,47 kg.

A utilização do Material B em substituição ao Material A proporciona uma redução de massa de aproximadamente 135,52 kg, correspondente a 23,69% da massa calculada para o Material A.

Esse resultado demonstra que um material de maior massa específica pode resultar em uma estrutura de menor massa quando suas propriedades mecânicas permitem reduzir suficientemente a espessura necessária.

O Material C apresentou o menor volume de matéria-prima entre as alternativas analisadas, correspondente a aproximadamente 0,045427 m³.

Considerando sua massa específica de 7900 kg/m³, obteve-se uma massa de aproximadamente 358,88 kg.

Comparativamente ao Material B, a utilização do Material C proporciona uma redução de massa de aproximadamente 77,59 kg, equivalente a 17,78%.

Em relação ao Material A, essa redução corresponde a aproximadamente 213,11 kg, equivalente a 37,26% da massa originalmente calculada.

Portanto, considerando exclusivamente a minimização da massa da parede cilíndrica, o Material C representa a alternativa mais vantajosa.

A redução da massa de um equipamento industrial pode apresentar vantagens relacionadas ao transporte, à movimentação, à instalação e às cargas transmitidas aos elementos de sustentação.

Entretanto, essas possíveis vantagens não foram quantificadas economicamente neste estudo.

A análise realizada considera exclusivamente a massa da parede cilíndrica, não incluindo os tampos, flanges, bocais, suportes, componentes internos e demais elementos que constituem o reator completo.

Consequentemente, os valores apresentados não deverão ser interpretados como as massas totais do equipamento industrial.

### 5. Análise econômica das alternativas

A avaliação econômica demonstrou que a alternativa de menor massa não corresponde necessariamente à alternativa de menor custo de matéria-prima.

O Material A apresentou uma massa de aproximadamente 571,99 kg e um custo unitário de R$ 8,00/kg, resultando em um custo estimado de matéria-prima de R$ 4.575,91.

Apesar de apresentar a maior massa entre as alternativas avaliadas, seu reduzido custo unitário permitiu obter o menor custo de matéria-prima.

O Material B apresentou uma massa de aproximadamente 436,47 kg e um custo unitário de R$ 24,00/kg, resultando em um custo estimado de R$ 10.475,17.

Comparativamente ao Material A, essa alternativa apresenta um acréscimo de aproximadamente R$ 5.899,25, correspondente a 128,92% do custo de matéria-prima do Material A.

O Material C, embora tenha apresentado a menor massa da parede cilíndrica, correspondente a aproximadamente 358,88 kg, possui um custo unitário de R$ 35,00/kg.

Consequentemente, seu custo estimado de matéria-prima corresponde a R$ 12.560,68, representando o maior custo entre as alternativas avaliadas.

A comparação entre os Materiais B e C permite compreender de maneira particularmente clara a influência simultânea da massa e do custo unitário sobre o custo de matéria-prima.

Embora o Material C proporcione uma redução de massa de aproximadamente 17,78% em relação ao Material B, seu custo unitário é aproximadamente 45,83% superior.

A redução da quantidade de matéria-prima não foi suficiente para compensar o aumento do preço de aquisição por unidade de massa.

Como consequência, o custo de matéria-prima do Material C é aproximadamente 19,91% superior ao custo do Material B.

A utilização do Material B em substituição ao Material C proporciona uma economia de aproximadamente R$ 2.085,52, correspondente a 16,60% do custo de matéria-prima do Material C.

Os resultados demonstram que a avaliação econômica deve considerar simultaneamente as propriedades mecânicas, as características geométricas resultantes do dimensionamento, as massas específicas e os custos unitários dos materiais.

A análise realizada permite estabelecer uma classificação econômica preliminar, na qual o Material A apresenta o menor custo, seguido pelos Materiais B e C.

Entretanto, essa classificação ainda deverá ser submetida às restrições técnicas estabelecidas para o problema.

Ressalta-se que os custos calculados correspondem exclusivamente à matéria-prima necessária à construção da parede cilíndrica, não representando o custo total de fabricação, instalação, operação ou manutenção do equipamento.

As diferenças monetárias e percentuais foram determinadas utilizando os resultados não arredondados da memória de cálculo.

### 6. Avaliação das restrições técnicas e admissibilidade dos materiais

A seleção de materiais para a construção de equipamentos industriais não pode ser fundamentada exclusivamente em critérios econômicos.

A viabilidade de uma alternativa depende inicialmente do atendimento aos requisitos técnicos estabelecidos para o projeto.

No presente estudo, os três materiais apresentaram espessuras comerciais adequadas e atenderam ao critério mecânico simplificado.

O Material A apresentou uma tensão circunferencial de 112,50 MPa, inferior à tensão admissível de 120,00 MPa, além de possuir o menor custo de matéria-prima, correspondente a R$ 4.575,91.

Entretanto, de acordo com as informações fornecidas nos dados de entrada, esse material apresenta incompatibilidade química com o fluido de processo.

Essa incompatibilidade constitui uma restrição eliminatória, independentemente de suas vantagens mecânicas e econômicas.

Em um equipamento industrial real, a incompatibilidade entre o material de construção e o meio de operação pode favorecer mecanismos de degradação capazes de comprometer a integridade do equipamento, reduzir sua vida útil e aumentar os riscos associados à operação.

Como não foram fornecidas informações detalhadas sobre a composição química do fluido ou os mecanismos de degradação envolvidos, não é possível identificar ou quantificar esses fenômenos neste estudo.

A exclusão do Material A fundamenta-se, portanto, na restrição de compatibilidade química estabelecida nos dados do problema.

Os Materiais B e C, por sua vez, apresentam compatibilidade química com o fluido de processo e atendem ao critério mecânico simplificado, sendo considerados tecnicamente admissíveis nas condições analisadas.

Dessa maneira, a seleção econômica final deverá ser realizada exclusivamente entre esses dois materiais.

O Material B apresenta um custo de matéria-prima de R$ 10.475,17, enquanto o Material C apresenta um custo de R$ 12.560,68.

Portanto, embora o Material A seja a alternativa de menor custo entre todos os materiais candidatos, o Material B representa a alternativa de menor custo entre os materiais tecnicamente admissíveis.

Essa distinção demonstra que as restrições técnicas não constituem critérios secundários de comparação, mas condições obrigatórias para a seleção de uma alternativa de engenharia.

Uma solução economicamente atrativa não poderá ser considerada adequada quando não satisfizer os requisitos técnicos estabelecidos para o problema.

### 7. Recomendação final de engenharia

Com base nos resultados obtidos durante o dimensionamento preliminar, a verificação mecânica, a avaliação das massas, a comparação econômica e a análise das restrições técnicas, recomenda-se preliminarmente a utilização do Material B para a construção da parede cilíndrica do reator químico.

A alternativa selecionada apresenta uma espessura comercial de 6,00 mm, um diâmetro externo de 1,212 m, um volume de matéria-prima de aproximadamente 0,054558 m³ e uma massa estimada de 436,47 kg.

Considerando o custo unitário de R$ 24,00/kg, o custo estimado de matéria-prima corresponde a R$ 10.475,17.

A tensão circunferencial calculada para essa configuração corresponde a 150,00 MPa, atendendo ao limite de tensão admissível estabelecido para o Material B.

Além disso, o material apresenta compatibilidade química com o fluido de processo, satisfazendo as restrições técnicas consideradas no estudo.

Embora o Material C proporcione uma redução de aproximadamente 77,59 kg na massa da parede cilíndrica, seu custo de matéria-prima é aproximadamente R$ 2.085,52 superior ao do Material B.

Consequentemente, a utilização do Material B proporciona uma economia de aproximadamente 16,60% em relação ao custo de matéria-prima do Material C.

O Material A, apesar de apresentar o menor custo absoluto e uma folga adicional de tensão de 7,50 MPa, foi excluído em decorrência de sua incompatibilidade química.

A recomendação do Material B não resulta, portanto, da minimização isolada da massa, da maximização da resistência mecânica ou da seleção do menor custo unitário.

Resulta da avaliação integrada das alternativas e da minimização do custo de matéria-prima entre os materiais que satisfazem as restrições técnicas estabelecidas.

Entretanto, deve-se observar que a espessura comercial selecionada coincide com a espessura mínima calculada e que a tensão circunferencial atinge exatamente o limite admissível fornecido.

Essa condição deverá receber atenção especial durante o desenvolvimento das etapas posteriores do projeto.

A espessura de 6,00 mm deverá ser considerada uma referência de dimensionamento preliminar, sujeita às verificações complementares necessárias para estabelecer a espessura definitiva de fabricação.

Dessa maneira, recomenda-se o Material B como a alternativa preliminarmente mais vantajosa segundo os critérios técnicos e econômicos considerados neste estudo, sem que essa recomendação constitua uma autorização para fabricação ou operação do equipamento.

### 8. Limitações do estudo e recomendações para o desenvolvimento do projeto

O presente estudo foi desenvolvido utilizando um modelo matemático simplificado destinado à análise preliminar de um casco cilíndrico submetido a uma diferença de pressão.

A metodologia permitiu comparar alternativas de materiais e estabelecer uma recomendação fundamentada nas propriedades e nas restrições fornecidas.

Entretanto, o modelo utilizado não contempla todos os fenômenos físicos, mecanismos de degradação e requisitos normativos associados ao projeto de um vaso de pressão industrial.

A análise mecânica foi realizada considerando exclusivamente a tensão circunferencial de uma parede cilíndrica de espessura uniforme, submetida a uma diferença de pressão de projeto de 1,50 MPa.

Não foram avaliadas detalhadamente as tensões longitudinais, as concentrações de tensão, as descontinuidades geométricas, os efeitos térmicos, as condições de fadiga e os demais carregamentos que poderão atuar sobre o equipamento.

Também não foram consideradas as influências das juntas soldadas, das tolerâncias dimensionais de fabricação, das possíveis perdas de espessura decorrentes de mecanismos de degradação e das características estruturais dos tampos, flanges, bocais e suportes.

A compatibilidade química foi tratada como uma informação previamente estabelecida, não tendo sido realizada uma investigação específica sobre o comportamento dos materiais nas condições reais de composição química, temperatura e operação do fluido de processo.

Do ponto de vista econômico, foram considerados exclusivamente os custos de matéria-prima da parede cilíndrica.

Não foram incorporados os custos de fabricação, conformação, soldagem, inspeção, montagem, transporte, instalação, manutenção e operação.

Consequentemente, a alternativa de menor custo de matéria-prima não representa necessariamente a alternativa de menor custo global durante a vida útil do equipamento.

Os materiais analisados e suas respectivas propriedades constituem dados hipotéticos fornecidos para o desenvolvimento deste exercício.

Antes da seleção definitiva de um material para aplicação industrial, será necessário identificar sua especificação comercial, verificar suas propriedades nas condições reais de operação e avaliar sua conformidade com os códigos e normas técnicas aplicáveis.

Considerando que o Material B apresentou uma tensão circunferencial coincidente com sua tensão admissível, recomenda-se que as etapas posteriores do projeto investiguem a necessidade de uma espessura nominal superior à espessura mínima calculada.

Caso requisitos adicionais de projeto conduzam à alteração das espessuras necessárias, os cálculos mecânicos, volumétricos, mássicos e econômicos deverão ser novamente realizados para todas as alternativas relevantes.

Essa reavaliação é necessária porque alterações nas condições de dimensionamento podem modificar a classificação econômica e, consequentemente, a seleção do material.

Portanto, os resultados apresentados deverão ser utilizados como subsídio para o desenvolvimento das etapas posteriores de engenharia, não constituindo uma aprovação definitiva do projeto para fabricação ou operação.

### 9. Síntese conclusiva do relatório técnico

O estudo preliminar de seleção de materiais demonstrou que a definição do material de construção de um equipamento industrial constitui um problema de engenharia que envolve simultaneamente aspectos mecânicos, físicos, geométricos, econômicos e químicos.

Os resultados permitiram identificar três características distintas entre as alternativas analisadas.

O Material A apresentou o menor custo de matéria-prima, correspondente a R$ 4.575,91, porém foi considerado não admissível em decorrência de sua incompatibilidade química.

O Material C apresentou a menor massa da parede cilíndrica, correspondente a 358,88 kg, porém resultou no maior custo de matéria-prima, correspondente a R$ 12.560,68.

O Material B apresentou uma condição intermediária de massa e custo, atendendo simultaneamente aos critérios mecânico e químico estabelecidos.

Sua utilização permite obter uma parede cilíndrica com espessura comercial de 6,00 mm, massa estimada de 436,47 kg e custo de matéria-prima de R$ 10.475,17.

Consequentemente, o Material B representa a solução do problema de otimização preliminar formulado neste estudo.

A economia de aproximadamente R$ 2.085,52 em relação ao Material C demonstra quantitativamente a vantagem econômica da alternativa recomendada entre os materiais admissíveis.

Entretanto, a ausência de folga adicional entre a tensão circunferencial calculada e a tensão admissível, associada às limitações do modelo simplificado, evidencia a necessidade de análises complementares antes da definição final das características construtivas do equipamento.

Conclui-se que a utilização integrada da modelagem matemática e da programação computacional permitiu estabelecer uma metodologia sistemática, transparente e reproduzível para a comparação de alternativas e a tomada de decisão em engenharia.

O desenvolvimento deste estudo permitiu compreender que a Engenharia Auxiliada por Computador não se limita à utilização de ferramentas de desenho, modelagem tridimensional ou programação computacional.

Essas ferramentas representam recursos que, quando integrados aos conhecimentos científicos e tecnológicos da engenharia, permitem compreender problemas, investigar alternativas e fundamentar decisões.

Inicialmente, a representação tridimensional do reator permitiu compreender sua configuração geométrica e identificar os elementos que constituem o equipamento.

Posteriormente, a utilização dos princípios da mecânica permitiu estabelecer um modelo matemático para o dimensionamento preliminar da parede cilíndrica.

A partir desse modelo, foram desenvolvidos os procedimentos de cálculo necessários à determinação das espessuras, das tensões circunferenciais, dos volumes, das massas e dos custos de matéria-prima.

A implementação computacional permitiu automatizar esses procedimentos, organizar os resultados, avaliar diferentes alternativas e investigar a influência das propriedades dos materiais sobre a seleção final.

Entretanto, a obtenção dos resultados numéricos não encerrou o problema de engenharia.

Foi necessário interpretar os resultados, compreender as relações existentes entre as grandezas calculadas, identificar as restrições técnicas e reconhecer as limitações dos modelos utilizados.

Essa sequência permitiu transformar um conjunto de informações geométricas, físicas, mecânicas e econômicas em uma decisão de engenharia fundamentada.

Ao longo deste estudo, verificamos que:

- O material de menor massa específica não necessariamente proporciona a estrutura de menor massa;
- O material de maior tensão admissível não necessariamente representa a alternativa economicamente mais vantajosa;
- O material de menor custo não necessariamente atende aos requisitos técnicos do projeto;
- O atendimento a um modelo matemático simplificado não equivale à aprovação definitiva de um equipamento industrial;
- A qualidade de uma decisão de engenharia depende tanto da qualidade dos modelos utilizados quanto da capacidade de interpretar seus resultados.

Essas observações demonstram que a utilização de recursos computacionais deve estar associada à compreensão dos fenômenos físicos, ao domínio dos fundamentos matemáticos e à capacidade de análise crítica.

O objetivo da Engenharia Auxiliada por Computador não consiste simplesmente em produzir desenhos, executar programas ou obter resultados numéricos.

Consiste em utilizar esses recursos para ampliar a capacidade do engenheiro de compreender, projetar, analisar e tomar decisões tecnicamente fundamentadas.

O computador executa os procedimentos matemáticos estabelecidos pelo engenheiro. Entretanto, a responsabilidade pela definição do problema, pela escolha do modelo, pela avaliação dos resultados e pela decisão final permanece com o profissional de engenharia.
